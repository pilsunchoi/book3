### SU

```python
def SU_pdf(x, m, s, lambda_, theta):

    J = 1/(theta*np.sqrt((x-m)**2+s**2))
    z = (np.arcsinh((x - m)/s) - lambda_)/theta
    pdf_vals = J*st.norm.pdf(z)
    
    return pdf_vals
```

```python
def SU_bi_pdf(x1, x2, m1, m2, s1, s2, lambda1, lambda2, theta1, theta2, r):

    J = 1/( (theta1*np.sqrt((x1-m1)**2+s1**2))*(theta2*np.sqrt((x2-m1)**2+s2**2)) )
    z1 = (np.arcsinh((x1 - m1)/s1) - lambda1)/theta1
    z2 = (np.arcsinh((x2 - m2)/s2) - lambda2)/theta2
    pdf_vals = 1/(2*np.pi*np.sqrt(1-r**2))*J*np.exp(
        -1/(2*(1-r**2))*(z1**2+z2**2-2*r*z1*z2))
    
    return pdf_vals
```

```python
def SU_cond_pdf(x1, x2, m1, m2, s1, s2, lambda1, lambda2, theta1, theta2, r):
    '''
    Conditional PDF of x1 given x2
    '''    
    
    lambda1_star = lambda1 + r * theta1/theta2 * (np.arcsinh((x2 - m2) / s2) - lambda2)
    theta1_star = theta1 * (1 - r**2)**(1/2)

    J = 1/(theta1_star*np.sqrt((x1-m1)**2+s1**2))
    z = (np.arcsinh((x1 - m1)/s1) - lambda1_star)/theta1_star
    pdf_vals = J*st.norm.pdf(z)
    
    return pdf_vals
```

#### Log Likelihood Functions

`s`($s$)와 `theta`($\theta$)가 항상 양수이기 때문에 최적화 과정에서 이 조건을 부과하기 위해 이들 파라미터에 지수(natural exponential)를 취해서 로그가능도 함수를 만드는 방법을 사용한다. 이렇게 하면 입력 파라미터는 제약(bounding) 조건이 사라지고 모든 실수값을 취할 수 있게 된다.

**일변량**

```python
def ll_SU_e(x, m, s, lambda_, theta):

    s = np.exp(s)
    theta = np.exp(theta)
    
    J = 1/(theta*np.sqrt((x-m)**2+s**2))
    z = (np.arcsinh((x - m)/s) - lambda_)/theta

    ln_pdf_vals = (np.log(J) - 0.5*np.log(2*np.pi) - 0.5*(z)**2)
    log_lik_val = ln_pdf_vals.sum()
    
    return log_lik_val
```

**이변량**

```python
def  ll_bi_SU_e(x1, x2, m1, m2, s1, s2, lambda1, lambda2, theta1, theta2, r):

    s1 = np.exp(s1)
    theta1 = np.exp(theta1)
    s2 = np.exp(s2)
    theta2 = np.exp(theta2)
    
    J = 1/( (theta1*np.sqrt((x1-m1)**2+s1**2))*(theta2*np.sqrt((x2-m2)**2+s2**2)) )
    z1 = (np.arcsinh((x1 - m1)/s1) - lambda1)/theta1
    z2 = (np.arcsinh((x2 - m2)/s2) - lambda2)/theta2
    
    ln_pdf_vals = ( np.log(J) - np.log(2*np.pi) - 0.5*np.log(1-r**2)
                   - 0.5/(1-r**2)*(z1**2+z2**2-2*r*z1*z2) )
    log_lik_val = ln_pdf_vals.sum()
    
    return log_lik_val
```

### MLE Objective Functions

**일변량**

```python
def obj_SU_e(params, *args):

    m, s, lambda_, theta = params
    x = args
    
    log_lik_val = ll_SU_e(x, m, s, lambda_, theta)
    neg_log_lik_val = -log_lik_val
    
    return neg_log_lik_val
```

**이변량**

```python
def obj_bi_SU_e(params, *args):

    m1, s1, lambda1, theta1, m2, s2, lambda2, theta2, r = params
    x1, x2 = args
    log_lik_val = ll_bi_SU_e(x1, x2, m1, m2, s1, s2, lambda1, lambda2, theta1, theta2, r)
    neg_log_lik_val = -log_lik_val
    
    return neg_log_lik_val
```

```python
# Skewed Normal Random Numbers
import numpy as np

N = 100000  # 샘플 수
np.random.seed(123)

delta1 = 0.5
delta2 = 0.3
r0 = 0.4

# 상관 행렬 정의 
R_star = np.array([
    [1.0,  delta1,  delta2],    
    [delta1,  1.0,  r0],
    [delta2,  r0,  1.0]
])

# Cholesky 분해를 사용한 변환 행렬 생성
L = np.linalg.cholesky(R_star)

# 3개의 독립 표준 정규 난수 생성
Z = np.random.randn(N, 3)

# 상관 구조를 반영한 난수 생성
y = Z @ L.T  # y0, y1, y2가 포함된 (N, 3) 행렬

# y0 기준으로 조건을 적용하여 z1, z2 생성
z1 = np.where(y[:, 0] >= 0, y[:, 1], -y[:, 1])
z2 = np.where(y[:, 0] >= 0, y[:, 2], -y[:, 2])

print(z1[:5], z2[:5])
```

```python
# Skewed-t Random Numbers

# (1) Skewed Normal
import numpy as np
import scipy.stats as st

N = 100000  # 샘플 수
np.random.seed(123)

delta1 = 0.5
delta2 = 0.3
r0 = 0.4

# 상관 행렬 정의 
R_star = np.array([
    [1.0,  delta1,  delta2],    
    [delta1,  1.0,  r0],
    [delta2,  r0,  1.0]
])

# Cholesky 분해를 사용한 변환 행렬 생성
L = np.linalg.cholesky(R_star)

# 3개의 독립 표준 정규 난수 생성
Z = np.random.randn(N, 3)

# 상관 구조를 반영한 난수 생성
y = Z @ L.T  # y0, y1, y2가 포함된 (N, 3) 행렬

# y0 기준으로 조건을 적용하여 z1, z2 생성
z1 = np.where(y[:, 0] >= 0, y[:, 1], -y[:, 1])
z2 = np.where(y[:, 0] >= 0, y[:, 2], -y[:, 2])

# (2) Skewed-t

nu0 = 5
ru = np.random.rand(N)
chi = st.chi2.ppf(ru, df=nu0)

x1 = z1 / np.sqrt(chi / nu0)
x2 = z2 / np.sqrt(chi / nu0)

print(x1[:5], x2[:5])
```

```python
# SU Random Numbers
import numpy as np

N = 100000
np.random.seed(123)

# 상호독립인 표준정규 난수 생성
z0 = np.random.randn(N)
z1 = np.random.randn(N)
    
# 상관계수가 r0인 표준정규 난수 생성
r0 = 0.5
z1 = z1
z2 = r0 * z1 + np.sqrt(1 - r0**2) * z0

m1, s1, lambda1, theta1 = 0, 1, 1, 1
m2, s2, lambda2, theta2 = 0, 1, -1, 1

x1 = m1 + s1 * np.sinh(lambda1 + theta1 * z1)
x2 = m2 + s2 * np.sinh(lambda2 + theta2 * z2)

print(x1[:5], x2[:5])
```

```python
##############
## SU Model ##
##############

m1_init, s1_init, lambda1_init, theta1_init = 0, 0, 0, 0
m2_init, s2_init, lambda2_init, theta2_init = 0, 0, 0, 0
r = 0

params_init = np.array([m1_init, s1_init, lambda1_init, theta1_init,
                       m2_init, s2_init, lambda2_init, theta2_init, r])
mle_args = (x1,x2)
res_bi_SU = opt.minimize(obj_bi_SU_e, params_init, args=mle_args,
                         method='L-BFGS-B',
                         bounds=[(None, None)] * 8 + [(-.99, .99)])

(m1_MLE, s1_MLE, lambda1_MLE, theta1_MLE,
 m2_MLE, s2_MLE, lambda2_MLE, theta2_MLE, r_MLE)=res_bi_SU.x

s1_MLE = np.exp(s1_MLE)
theta1_MLE = np.exp(theta1_MLE)
s2_MLE = np.exp(s2_MLE)
theta2_MLE = np.exp(theta2_MLE)

result = {'m1': [m1_MLE], 's1': [s1_MLE], 
          'lambda1': [lambda1_MLE], 'theta1': [theta1_MLE],
          'm2': [m2_MLE], 's2': [s2_MLE],
          'lambda2': [lambda2_MLE], 'theta2': [theta2_MLE], 'r': [r_MLE]}
df = pd.DataFrame(result)

###################

# 시뮬레이션 설정
np.random.seed(456)
nn = 1000000
z0 = np.random.randn(nn)
z1 = np.random.randn(nn)
y1 = z1
y2 = r_MLE * z1 + np.sqrt(1 - r_MLE**2) * z0

su1 = m1_MLE + s1_MLE * np.sinh(lambda1_MLE + theta1_MLE * y1)
su2 = m2_MLE + s2_MLE * np.sinh(lambda2_MLE + theta2_MLE * y2)

w1 = 0.5
w2 = 1 - w1
sim = w1 * su1 + w2 * su2

# VaR 검정 수행
ind = np.zeros((N, 6))
port = w1 * x1 + w2 * x2
sim1 = np.sort(sim, axis=0)

q_l = np.array([0.005, 0.01, 0.05])
q_h = np.array([0.95, 0.99, 0.995])
v = np.hstack((q_l, q_h))
p = np.hstack((q_l, 1 - q_h))

VaR = sim1[(nn * v).astype(int)] 

ind[:, 0] = (port.flatten() < VaR[0])
ind[:, 1] = (port.flatten() < VaR[1])
ind[:, 2] = (port.flatten() < VaR[2])
ind[:, 3] = (port.flatten() > VaR[3])
ind[:, 4] = (port.flatten() > VaR[4])
ind[:, 5] = (port.flatten() > VaR[5])

ntest = len(port)

nviolation = np.sum(ind, axis=0)
er = nviolation / (ntest * p)

# 로그 가능도 비율 검정
a = nviolation / ntest
L_A = nviolation * np.log(a) + (ntest - nviolation) * np.log(1 - a)
L_0 = (nviolation * np.log(p) + (ntest - nviolation) * np.log(1 - p))
LR = 2 * (L_A - L_0)
LR_prob = 1 - st.chi2.cdf(LR, df=1)

# 결과 출력
print("SU MODEL:")
print("\nSimulation Parameters:")
print(N, r0, gam1, gam2, nu0)
print("\nOptimized Log Likelihood:", -res_bi_SU.fun)
print("\nParameter Estimates:")
print(df.iloc[:, :4])
print(df.iloc[:, 4:])
print("\nVaR Exceeding Ratio Results:")
print(er)
star1 = np.where(LR_prob < 0.01, 1, 0)
star2 = np.where(LR_prob < 0.05, 1, 0)
star = star1 + star2
print("\n* 5%, ** 1% (수준에서 1과 다름):")
print(star)
```
