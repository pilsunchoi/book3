## 요약  
본 논문에서는 다변량 확장 비대칭 t (EST) 분포 계열을 소개하고, 확장 비대칭 정규 분포와 같은 밀접한 관련 계열과 함께 자세히 연구한다. EST 분포는 수학적으로 다루기 용이하고, 왜도 및 정규 분포보다 두터운 꼬리를 모델링할 수 있는 유연성을 제공하는 것 외에도, 조건부 분포에 대한 닫힘성(closed under conditioning)과 가벼운 꼬리도 모델링할 수 있는 능력을 갖춘다는 점에서 중요한 특징을 가진다.  

논문의 첫 번째 부분에서는 EST 분포의 확률적 성질을 검토하며, 다양한 확률적 표현, 주변 및 조건부 분포, 선형 변환, 모멘트, 그리고 특히 Mardia의 다변량 왜도 및 첨도 측정을 포함한다. 두 번째 부분에서는 EST 분포의 통계적 성질을 연구하며, 우도 추론, 프로파일 로그 우도의 거동, 스코어 벡터 및 피셔 정보 행렬을 분석한다. 특히, 확장 비대칭 정규 (ESN) 분포와 달리, 단변량 EST 분포의 피셔 정보 행렬은 왜도가 0으로 설정될 때 비특이적(non-singular)임을 보인다. 마지막으로, 조건부 EST 분포의 수치적 응용을 기밀 데이터 변형(confidential data perturbation) 맥락에서 제시한다.  

**주요 키워드**
기밀 데이터 변형(Confidential data perturbation); 타원형 분포(Elliptically contoured distributions); 피셔 정보(Fisher information); 첨도; 비정규성; 왜도  


## 1 서론

다변량 분포의 유연한 모수적 계열을 탐색하는 연구는 최근 지속적인 관심을 받아왔다. 이는 Genton(2004)이 편집한 책, Azzalini(2005)의 리뷰 및 그에 포함된 참고문헌에서도 확인할 수 있다. 고려되는 분포 계열은 특정 선택 메커니즘에서 비롯된 비대칭 분포(skewed distributions)의 통합적 틀 안에서 다룰 수 있지만, 다변량 비대칭 t-분포(skew-t distribution)는 데이터의 비대칭성과 첨도를 조절할 수 있는 다루기 쉬우면서도 강건한 모델로 자리 잡았다. Azzalini와 Genton(2008)은 여러 이론적 논거와 예시를 통해 이 분포의 사용을 주장한 바 있다. 그러나 비대칭 t-분포에는 몇 가지 한계가 있으며, 본 논문에서는 이를 해결하고자 한다.  

첫째, 비대칭 t-분포는 조건부 분포에 대해 닫혀 있지 않다. 이는 데이터에 적합된 비대칭 t-모델의 조건부 분포를 연구해야 하는 방법론에서 중요한 문제이다. 예를 들어, 다양한 데이터 변형(data perturbation) 기법에서는 기밀 변수를 비기밀 변수로 조건화한 조건부 분포에서 샘플링해야 하며, 이에 대한 최근 연구와 참고문헌은 Muralidhar와 Sarathy(2003)를 참조할 수 있다. 마찬가지로, 검열된 데이터(censored data)에 대한 일부 대체(imputation) 방법에서는 검열된 관측값을 조건부 분포에서 샘플링한 값으로 대체해야 하며, 검열된 시계열의 경우 Park et al.(2007)을 참고할 수 있다.  

둘째, 비대칭 t-분포는 정규 분포보다 더 두터운 꼬리를 가진 분포를 모델링할 수 있지만, 더 가벼운 꼬리는 표현할 수 없다. 이러한 문제들은 비대칭 t-분포의 확장된 버전을 다음과 같이 정의함으로써 해결할 수 있다.

### 정의 1 (확장된 비대칭 t-분포)  
연속적인 $p$-차원 확률 벡터 $Y$가 다변량 확장 비대칭 $t$ 분포(Extended Skew-t, EST)를 따른다고 할 때, 이를 $Y \sim EST_p(\xi, \Omega, \lambda, \nu, \tau)$로 표기하며, 그 **밀도 함수**는 $y \in \mathbb{R}^p$에서 다음과 같이 정의된다.  

$$
\frac{1}{T_1\left({\tau}/{\sqrt{1 + \lambda^T \bar{\Omega} \lambda}}; \nu \right)}
t_p(y; \xi, \Omega, \nu) \ T_1 \left\{ \left(\lambda^T z + \tau \right) \left(\frac{\nu + p}{\nu + Q(z)}\right)^{1/2} ; \nu + p \right\}, \tag{1}
$$

여기서 $z = \omega^{-1} (y - \xi)$, $Q(z) = z^T \bar{\Omega}^{-1} z$, $\lambda \in \mathbb{R}^p$는 형태(shape) 파라미터, $\tau \in \mathbb{R}$는 확장(extension) 파라미터이다. 또한, $t_p(y; \xi, \Omega, \nu)$는 $p$-차원 Student’s $t$ 분포의 밀도 함수로서 다음과 같다. 

$$
t_p(y; \xi, \Omega, \nu) = \frac{\Gamma((\nu + p)/2)}{|\Omega|^{1/2} (\pi \nu)^{p/2} \Gamma(\nu/2)}
\left( 1 + \frac{Q(z)}{\nu} \right)^{-(\nu + p)/2}
$$

여기서 $\xi \in \mathbb{R}^p$는 위치(location) 파라미터, $\Omega$는 양의 정부호 $p \times p$ 분산 행렬, $\omega = \text{diag}(\Omega)^{1/2}$는 $p \times p$ 스케일 행렬, $\bar{\Omega} = \omega^{-1} \Omega \omega^{-1}$는 상관 행렬, $\nu > 0$는 자유도(degrees of freedom)이다. 마지막으로, $T_1(y; \nu)$는 자유도 $\nu > 0$를 갖는 일변량 표준 Student’s $t$ 누적분포함수(CDF)를 나타낸다.

$\it{EST}$ 분포는 매우 유연한 통계적 모델 계열을 제공한다. 실제로, $\lambda = \tau = 0$일 때, 다변량 대칭 Student’s t-분포 $t_p(\xi, \Omega, \nu)$를 얻을 수 있으며, 자유도를 $\nu \to \infty$로 하면 다변량 정규 분포 $N_p(\xi, \Omega)$가, $\nu = 1$로 하면 다변량 코시 분포 $C_p(\xi, \Omega)$가 된다. 동일한 현상이 $\tau \to +\infty$일 때도 발생하며, 반면 $\tau \to -\infty$일 때는 밀도가 0으로 수렴한다.  

$\tau = 0$일 경우, Azzalini와 Capitanio(2003)가 정의한 다변량 비대칭 t-분포 $ST_p(\xi, \Omega, \nu, \lambda)$가 된다. 또한, $\nu \to \infty$로 하면 Azzalini와 Dalla Valle(1996)의 다변량 비대칭 정규 분포 $SN_p(\xi, \Omega, \lambda)$가, $\nu = 1$로 하면 Arnold와 Beaver(2000a)의 비대칭 코시 분포 $SC_p(\xi, \Omega, \lambda)$가 얻어진다. 특히 $\lambda = 0$일 때, 다음과 같은 새로운 대칭 밀도 계열을 얻을 수 있다.  

$$
\frac{1}{T_1\left(\tau; \nu \right)}
t_p(y; \xi, \Omega, \nu) \ T_1 \left\{ \tau \left(\frac{\nu + p}{\nu + Q(z)}\right)^{1/2} ; \nu + p \right\}, \tag{2}
$$

이는 $\xi + C Z$의 밀도와 대응되며, 여기서  $Z \overset{d}{=} (X | X_0 + \tau > 0)$, $(X^T, X_0)^T \sim t_{p+1}(0, I_{p+1}, \nu)$ (즉, $(p+1)$-차원 Student’s t-분포), $C = \Omega^{1/2}$ (즉, $\Omega$의 제곱근)이다. 마지막으로, $\nu \to \infty$일 경우 확장된 비대칭 정규 분포 $ESN_p(\xi, \Omega, \lambda, \tau)$를 얻으며, 그 밀도는 다음과 같다.  

$$
\frac{1}{\Phi_1 \left( \tau / \sqrt{1 + \lambda^T \bar{\Omega} \lambda} \right)}
\phi_p(y; \xi, \Omega) \Phi_1(\lambda^T z + \tau), \tag{3}
$$

이는 Azzalini와 Capitanio(1999), Arnold와 Beaver(2000b)에서도 약간 다른 파라미터화로 등장한 바 있다. $ESN$ 분포는 Capitanio et al.(2003)에 의해 더 자세히 연구되었으나, 특정한 $\tau$ 값에서 정규 분포보다 가벼운 꼬리를 가질 수 있다는 점은 지적되지 않았다. 본 논문에서는 이 흥미로운 성질을 분석한다. 마지막으로, Azzalini와 Capitanio(2003, Section 4.2.4)에서도 확장된 비대칭 $t$ 분포의 정의를 간략히 제시한 바 있으나, 여기서는 $T_1$ 대신 정규 누적분포함수 $\Phi_1$을 기반으로 정규화를 수행하였다. 그러나 이러한 확장은 (1)식에서 제시한 확장만큼 자연스럽지 않은 것으로 보인다.

확장된 비대칭 t-분포의 형태를 설명하기 위해 먼저 $p = 1$인 경우를 고려한다. 이때, $Y \sim EST_1(\xi, \omega, \lambda, \nu, \tau)$인 확률변수의 밀도 함수는 다음과 같다.  

$$
\frac{1}{T_1\left({\tau}/{\sqrt{1 + \lambda^2}}; \nu \right)}
t_1(y; \xi, \omega, \nu) \ T_1 \left\{ \left(\lambda z + \tau \right) \left(\frac{\nu + 1}{\nu + z^2}\right)^{1/2} ; \nu + 1 \right\}, \tag{4}
$$

여기서 $z = (y - \xi)/\omega$이고, $\omega > 0$, $\nu > 0$이다. $T_1(y; \nu)$는 자유도 $\nu > 0$를 갖는 일변량 표준 Student’s $t$ 누적분포함수(CDF)를 나타낸다. 또한, $t_1(y; \xi, \omega, \nu)$는 Student’s $t$ 분포의 밀도 함수로서 다음과 같다. 

$$
t_1(y; \xi, \omega, \nu) = \frac{\Gamma((\nu + 1)/2)}{(\pi \nu)^{1/2} \Gamma(\nu/2)}
\left( 1 + \frac{z^2}{\nu} \right)^{-(\nu + 1)/2}
$$

여기서 $z = (y - \xi)/\omega$이고, $\omega > 0$, $\nu > 0$이다. 

```python
import numpy as np
from scipy.special import gamma

def t1_density(y, xi, omega, nu):
    """
    Computes the density of the univariate Student's t-distribution:
    
        t1(y; xi, omega, nu) = (Γ((nu + 1)/2) / (sqrt(pi * nu) * Γ(nu/2))) 
                               * (1 + z^2 / nu)^(-(nu + 1)/2)

    where z = (y - xi) / omega
    """
    z = (y - xi) / omega
    coef = gamma((nu + 1) / 2) / (np.sqrt(np.pi * nu) * gamma(nu / 2))
    density = coef * (1 + z**2 / nu) ** (-(nu + 1) / 2)
    return density / omega  # scale adjustment for omega
```

```python
import numpy as np
from scipy.stats import t

def t1_density(y, xi, omega, nu):
    z = (y - xi) / omega
    coef = gamma((nu + 1) / 2) / (np.sqrt(np.pi * nu) * gamma(nu / 2))
    density = coef * (1 + z**2 / nu) ** (-(nu + 1) / 2)
    return density / omega

def est1_density(y, xi, omega, lam, nu, tau):
    """
    Computes the density of the univariate Extended Skew-t distribution EST1.
    
        f(y) = (1 / T1(tau / sqrt(1 + lambda^2); nu)) * t1_density(y; xi, omega, nu) *
               T1( (lambda * z + tau) * sqrt((nu + 1)/(nu + z^2)); nu + 1 )

    where z = (y - xi) / omega
    """
    z = (y - xi) / omega
    
    # denominator normalization term
    denom_cdf = t.cdf(tau / np.sqrt(1 + lam**2), df=nu)
    
    # t1 density
    t1 = t1_density(y, xi, omega, nu)
    
    # transformed argument for second CDF
    arg = (lam * z + tau) * np.sqrt((nu + 1) / (nu + z**2))
    cdf_term = t.cdf(arg, df=nu + 1)
    
    return (1 / denom_cdf) * t1 * cdf_term
```

```python
y_vals = np.linspace(-5, 5, 100)
density_vals = [est1_density(y, xi=0, omega=1, lam=2, nu=3, tau=1) for y in y_vals]

import matplotlib.pyplot as plt
plt.plot(y_vals, density_vals)
plt.title("EST1(0, 1, 2, 3, 1) Density")
plt.xlabel("y")
plt.ylabel("Density")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t
from scipy.special import gamma

def t1_density(y, xi, omega, nu):
    z = (y - xi) / omega
    coef = gamma((nu + 1) / 2) / (np.sqrt(np.pi * nu) * gamma(nu / 2))
    density = coef * (1 + z**2 / nu) ** (-(nu + 1) / 2)
    return density / omega

def est1_density(y, xi, omega, lam, nu, tau):
    z = (y - xi) / omega
    denom_cdf = t.cdf(tau / np.sqrt(1 + lam**2), df=nu)
    t1 = t1_density(y, xi, omega, nu)
    arg = (lam * z + tau) * np.sqrt((nu + 1) / (nu + z**2))
    cdf_term = t.cdf(arg, df=nu + 1)
    return (1 / denom_cdf) * t1 * cdf_term

# 설정
xi = 0
omega = 1
lam = 2
nu = 3
tau_values = [0, 1, 3, 5, -1, -3, -5]
y_vals = np.linspace(-3, 6, 500)

# 그래프 그리기
plt.figure(figsize=(10, 6), dpi = 500)

for tau in tau_values:
    density = [est1_density(y, xi, omega, lam, nu, tau) for y in y_vals]
    if tau == 0:
        plt.plot(y_vals, density, 'k-', linewidth=2.5, label=r'$\tau=0$')
    elif tau > 0:
        plt.plot(y_vals, density, linestyle='-', label=rf'$\tau={tau}$')
    else:
        plt.plot(y_vals, density, linestyle='--', label=rf'$\tau={tau}$')

plt.title(r'Densities of EST$_1$(0, 1, 2, 3, $\tau$)')
plt.xlabel('y')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t
from scipy.special import gamma

def t1_density(y, xi, omega, nu):
    z = (y - xi) / omega
    coef = gamma((nu + 1) / 2) / (np.sqrt(np.pi * nu) * gamma(nu / 2))
    density = coef * (1 + z**2 / nu) ** (-(nu + 1) / 2)
    return density / omega

def est1_density(y, xi, omega, lam, nu, tau):
    z = (y - xi) / omega
    denom_cdf = t.cdf(tau / np.sqrt(1 + lam**2), df=nu)
    t1 = t1_density(y, xi, omega, nu)
    arg = (lam * z + tau) * np.sqrt((nu + 1) / (nu + z**2))
    cdf_term = t.cdf(arg, df=nu + 1)
    return (1 / denom_cdf) * t1 * cdf_term

# 설정값
omega = 1
lam = 2
nu = 3
tau = 0
xi_values = [-2, 0, 2]
y_vals = np.linspace(-8, 8, 500)

# 그래프 그리기
plt.figure(figsize=(10, 6))

for xi in xi_values:
    density = [est1_density(y, xi, omega, lam, nu, tau) for y in y_vals]
    plt.plot(y_vals, density, label=rf'$\xi={xi}$')

plt.title(r'Densities of EST$_1(\xi, 1, 2, 3, 0)$ for different $\xi$')
plt.xlabel('y')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t
from scipy.special import gamma

def t1_density(y, xi, omega, nu):
    z = (y - xi) / omega
    coef = gamma((nu + 1) / 2) / (np.sqrt(np.pi * nu) * gamma(nu / 2))
    density = coef * (1 + z**2 / nu) ** (-(nu + 1) / 2)
    return density / omega

def est1_density(y, xi, omega, lam, nu, tau):
    z = (y - xi) / omega
    denom_cdf = t.cdf(tau / np.sqrt(1 + lam**2), df=nu)
    t1 = t1_density(y, xi, omega, nu)
    arg = (lam * z + tau) * np.sqrt((nu + 1) / (nu + z**2))
    cdf_term = t.cdf(arg, df=nu + 1)
    return (1 / denom_cdf) * t1 * cdf_term

# 고정된 파라미터
xi = 0
lam = 2
nu = 3
tau = 0
omega_values = [0.5, 1, 2]
y_vals = np.linspace(-8, 8, 500)

# 그래프 그리기
plt.figure(figsize=(10, 6))

for omega in omega_values:
    density = [est1_density(y, xi, omega, lam, nu, tau) for y in y_vals]
    plt.plot(y_vals, density, label=rf'$\omega={omega}$')

plt.title(r'Densities of EST$_1(0, \omega, 2, 3, 0)$ for different $\omega$')
plt.xlabel('y')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

그림 1은 $\xi = 0$, $\omega = 1$, $\lambda = 2$, $\nu = 3$일 때, $\tau$의 다양한 값에 따른 밀도 함수 (4)의 형태를 나타낸다. 두꺼운 실선은 $\tau = 0$에 해당하며, 이는 고전적인 $ST$ 밀도이다. $\tau = 1, 3, 5$ (얇은 실선)에서는 밀도가 더 대칭적으로 변하며, $t_1(y; \xi, \omega^2, \nu)$ 밀도에 가까워진다. $\tau = -1, -3, -5$ (점선)에서는 밀도가 더 비대칭적으로 변한다.  

이 논문에서는 확장된 비대칭 $t$ 분포의 모멘트를 다룬다. 예제로서, 그림 2에서는 단변량 표준 확장 비대칭 정규 분포, 즉 $ESN_1(0, 1, \lambda, \tau)$에 대한 평균 $\mu$, 분산 $\sigma^2$, 비대칭도 계수 $\sqrt{\beta_{1,1}}$, 첨도 계수 $\beta_{2,1}$을 $\tau$의 함수로 나타냈다.  두꺼운 곡선은 $\lambda = 1$에 해당한다. $\tau > 0$의 특정 값에서 첨도 $\beta_{2,1} < 0$이 되는 경우가 관찰되며, 이는 정규 분포보다 꼬리가 가벼운 분포를 의미한다.

- $\lambda = 1, 2, 3, 4, 5$에 대한 $\tau$의 함수  
- 두꺼운 곡선은 $\lambda = 1$을 나타냄



비록 정의 1의 확장된 비대칭 t-분포가 실용적인 응용에서 매력적이지만, 이론적으로 흥미로운 두 가지 자연스러운 일반화가 존재한다. 첫째, 연속적인 $p$차원 랜덤 벡터 $Y$가 다변량 통합 비대칭 $t$ (${\it SUT}$) 분포를 따른다고 할 때, 이를 $Y \sim {\it SUT}_{p,q}(\xi, \Omega, \Lambda, \nu, \tau, \Gamma)$로 표기하며, $y \in \mathbb{R}^p$에서의 밀도 함수는 다음과 같다.  

$$
\frac{1}{T_q(\tau; \Gamma + \Lambda \bar{\Omega} \Lambda^T, \nu)} \ 
t_p(y; \xi, \Omega, \nu) \ T_q \left\{ \left( \Lambda z + \tau \right) \left(\frac{\nu + p}{\nu + Q(z)}\right)^{1/2} ; \Gamma, \nu+p \right\}, \tag{5}
$$

여기서 $\Lambda$는 $q \times p$ 차원의 실수 행렬로 형태를 조절하며, $\tau \in \mathbb{R}^q$는 확장 매개변수, $\Gamma$는 $q \times q$ 차원의 양정정부호 상관 행렬이며, $T_r(y; \Theta, \nu)$는 $r$차원 중심화된 스튜던트 $t$ 누적 분포 함수로 $r \times r$ 양정정부호 분산 행렬 $\Theta$와 자유도 $\nu > 0$을 가진다. $q = 1$일 때, $\Gamma = 1$, $\Lambda = \lambda^T$, 그리고 $T_q(x; a, \nu) = T_1(x / \sqrt{a}; \nu)$이므로 (5)는 (1)로 축소되며, 즉 ${\it SUT}_{p,1} = {\it EST}_p$가 된다. $\nu \to \infty$이면 다음과 같은 통합 비대칭 정규(${\it SUN}$) 밀도로 축소된다.  

$$
\frac{1}{\Phi_q(\tau; \Gamma + \Lambda \bar{\Omega} \Lambda^T)} \ 
\phi_p(y; \xi, \Omega) \ \Phi_q (\Lambda z + \tau; \Gamma),
$$

이를 ${\it SUN}_{p,q}(\xi, \Omega, \Lambda, \tau, \Gamma)$로 표기하며, Arellano-Valle와 Azzalini (2006)이 다른 파라미터화(parameterization) 방식으로 연구하였다. $q > 1$인 경우, 다변량 누적 분포 함수 $\Phi_q$ 및 $T_q$를 계산하는 알고리즘이 존재하지만 (Genz, 1992; Genz & \Lambdaretz, 2002 참조), 실용적으로는 $q = 1$을 사용하는 것이 더 간편하다. 그러나 $p = q$이고 행렬 $\bar{\Omega}, \Lambda, \Gamma = I_p$가 대각 행렬이면, ${\it SUN}_{p,p}$ 분포는 $p$개의 ${\it ESN}_1$ 분포의 곱으로 축소되며, 이는 독립적이고 동일한 분포를 따르는 ${\it ESN}_1(\xi, \omega^2, \lambda, \tau)$ 변수들의 특수한 경우를 포함한다. 

${\it SUT}$와 ${\it SUN}$은 타원형 분포(${\it EC}$)에서 유래한 보다 일반적인 가족의 두 가지 중요한 부분집합에 불과하다 (Fang et al., 1990 참조). 이를 설명하기 위해 다음을 정의한다. 연속적인 $p$차원 랜덤 벡터 $Y$가 다변량 통합 비대칭 타원형 (${\it SUE}$) 분포를 따른다고 할 때, 이를 $Y \sim {SUE}_{p,q}(\xi, \Omega, \Lambda, h(p+q), \tau, \Gamma)$로 표기하며, $y \in \mathbb{R}^p$에서의 밀도 함수는 다음과 같다.  

$$
\frac{1}{F_q(\tau; \Gamma + \Lambda \bar{\Omega} \Lambda^T , h(q))}
f_p(y; \xi, \Omega, h(p)) F_q \left( \frac{\Lambda z + \tau}{\sqrt{Q(z)}}; \Gamma, h(q) \right),
$$

여기서  $f_p(y; \xi, \Omega, h(p)) = |\Omega|^{-1/2} h(p)(Q(z))$는 위치 벡터 $\xi \in \mathbb{R}^p$, 양정정부호 분산 행렬 $\Omega$, 밀도 생성 함수 $h(p)$를 가지는 타원형 분포의 밀도 함수이다. $F_r(y; \Theta, h(r))$는 $r$차원 중심화된 타원형 누적 분포 함수로 $r \times r$ 분산 행렬 $\Theta$와 밀도 생성 함수 $h(q)$를 가진다. $h(q)_{Q(z)}(u) = h(p+q)(u + Q(z))/h(p)(Q(z))$이다. 이 ${\it SUE}$ 분포는 Arellano-Valle와 Azzalini (2006)에서 다른 파라미터화 방식으로 연구되었으며, Arellano-Valle et al. (2006), Arellano-Valle & Genton (2005), González-Farías et al. (2004)에서도 언급되었다. 그러나 이 연구들은 ${\it SUE}$ 분포의 주변 및 조건부 분포, 모멘트 등의 주요 특성을 체계적으로 분석하지 않았다. 분산 행렬 $\Omega$의 인수분해를 제외하면, 우리는 González-Farías et al. (2004)의 폐합 비대칭 정규 분포와 유사한 파라미터화를 사용하고 있다. ${\it SUE}$ 분포는 $h(p)$가 자유도 $\nu$를 가진 $p$차원 스튜던트 $t$ 밀도 생성 함수일 때 $SUT$로 축소되며,  

$$
h^{(p)}(u) = c(\nu, p) \{1 + (u/\nu)\}^{-(\nu+p)/2},\text{ 여기서 }c(a, b) = \frac{\Gamma((a + b)/2)}{\Gamma(a/2)(\pi a)^{b/2}}. \tag{6}
$$

또한 $q = 1$인 경우, $\it SUE$ 분포를 확장된 비대칭 타원형 ($\it ESE$) 분포라고 부르며, 이는 정의 1과의 유사성에서 비롯된다. 이들은 (6)의 스튜던트 $t$ 밀도 생성 함수를 채택하면 직접적으로 $\it EST$ 분포로 축소된다. 

비록 정의 1의 확장된 비대칭 t-분포가 실용적인 응용에서 매력적이지만, 이론적으로 흥미로운 두 가지 자연스러운 일반화가 존재한다.  

첫째, 연속적인 $p$차원 랜덤 벡터 $Y$가 다변량 통합 비대칭 $t$ (${\it SUT}$) 분포를 따른다고 할 때, 이를 $Y \sim {\it SUT}_{p,q}(\xi, \Omega, \Lambda, \nu, \tau, \Gamma)$로 표기하며, $y \in \mathbb{R}^p$에서의 밀도 함수는 다음과 같다.  

$$
\frac{1}{T_q(\tau; \Gamma + \Lambda \bar{\Omega} \Lambda^T, \nu)} \ 
t_p(y; \xi, \Omega, \nu) \ T_q \left\{ \left( \Lambda z + \tau \right) \left(\frac{\nu + p}{\nu + Q(z)}\right)^{1/2} ; \Gamma, \nu+p \right\}, \tag{5}
$$

여기서 $\Lambda$는 $q \times p$ 차원의 실수 행렬로 형태를 조절하며, $\tau \in \mathbb{R}^q$는 확장 매개변수, $\Gamma$는 $q \times q$ 차원의 양정정부호 상관 행렬이며, $T_r(y; \Theta, \nu)$는 $r$차원 중심화된 스튜던트 $t$ 누적 분포 함수로 $r \times r$ 양정정부호 분산 행렬 $\Theta$와 자유도 $\nu > 0$을 가진다. $q = 1$일 때, $\Gamma = 1$, $\Lambda = \lambda^T$, 그리고 $T_q(x; a, \nu) = T_1(x / \sqrt{a}; \nu)$이므로 (5)는 (1)로 축소되며, 즉 ${\it SUT}_{p,1} = {\it EST}_p$가 된다. $\nu \to \infty$이면 다음과 같은 통합 비대칭 정규(${\it SUN}$) 밀도로 축소된다.  

$$
\frac{1}{\Phi_q(\tau; \Gamma + \Lambda \bar{\Omega} \Lambda^T)} \ 
\phi_p(y; \xi, \Omega) \ \Phi_q (\Lambda z + \tau; \Gamma),
$$

이를 ${\it SUN}_{p,q}(\xi, \Omega, \Lambda, \tau, \Gamma)$로 표기하며, Arellano-Valle와 Azzalini (2006)이 다른 파라미터화(parameterization) 방식으로 연구하였다. $q > 1$인 경우, 다변량 누적 분포 함수 $\Phi_q$ 및 $T_q$를 계산하는 알고리즘이 존재하지만 (Genz, 1992; Genz & \Lambdaretz, 2002 참조), 실용적으로는 $q = 1$을 사용하는 것이 더 간편하다. 그러나 $p = q$이고 행렬 $\bar{\Omega}, \Lambda, \Gamma = I_p$가 대각 행렬이면, ${\it SUN}_{p,p}$ 분포는 $p$개의 ${\it ESN}_1$ 분포의 곱으로 축소되며, 이는 독립적이고 동일한 분포를 따르는 ${\it ESN}_1(\xi, \omega^2, \lambda, \tau)$ 변수들의 특수한 경우를 포함한다. ${\it SUT}$와 ${\it SUN}$은 타원형 분포(${\it EC}$)에서 유래한 보다 일반적인 가족의 두 가지 중요한 부분집합에 불과하다 (Fang et al., 1990 참조). 이를 설명하기 위해 다음을 정의한다. 연속적인 $p$차원 랜덤 벡터 $Y$가 다변량 통합 비대칭 타원형 (${\it SUE}$) 분포를 따른다고 할 때, 이를 $Y \sim {SUE}_{p,q}(\xi, \Omega, \Lambda, h(p+q), \tau, \Gamma)$로 표기하며, $y \in \mathbb{R}^p$에서의 밀도 함수는 다음과 같다.  

$$
\frac{1}{F_q(\tau; \Gamma + \Lambda \bar{\Omega} \Lambda^T , h(q))}
f_p(y; \xi, \Omega, h(p)) F_q \left( \frac{\Lambda z + \tau}{\sqrt{Q(z)}}; \Gamma, h(q) \right),
$$

여기서  $f_p(y; \xi, \Omega, h(p)) = |\Omega|^{-1/2} h(p)(Q(z))$는 위치 벡터 $\xi \in \mathbb{R}^p$, 양정정부호 분산 행렬 $\Omega$, 밀도 생성 함수 $h(p)$를 가지는 타원형 분포의 밀도 함수이다. $F_r(y; \Theta, h(r))$는 $r$차원 중심화된 타원형 누적 분포 함수로 $r \times r$ 분산 행렬 $\Theta$와 밀도 생성 함수 $h(q)$를 가진다. $h(q)_{Q(z)}(u) = h(p+q)(u + Q(z))/h(p)(Q(z))$이다. 이 ${\it SUE}$ 분포는 Arellano-Valle와 Azzalini (2006)에서 다른 파라미터화 방식으로 연구되었으며, Arellano-Valle et al. (2006), Arellano-Valle & Genton (2005), González-Farías et al. (2004)에서도 언급되었다. 그러나 이 연구들은 ${\it SUE}$ 분포의 주변 및 조건부 분포, 모멘트 등의 주요 특성을 체계적으로 분석하지 않았다. 분산 행렬 $\Omega$의 인수분해를 제외하면, 우리는 González-Farías et al. (2004)의 폐합 비대칭 정규 분포와 유사한 파라미터화를 사용하고 있다. ${\it SUE}$ 분포는 $h(p)$가 자유도 $\nu$를 가진 $p$차원 스튜던트 $t$ 밀도 생성 함수일 때 $SUT$로 축소되며,  

$$
h^{(p)}(u) = c(\nu, p) \{1 + (u/\nu)\}^{-(\nu+p)/2},\text{ 여기서 }c(a, b) = \frac{\Gamma((a + b)/2)}{\Gamma(a/2)(\pi a)^{b/2}}. \tag{6}
$$

또한 $q = 1$인 경우, $\it SUE$ 분포를 확장된 비대칭 타원형 ($\it ESE$) 분포라고 부르며, 이는 정의 1과의 유사성에서 비롯된다. 이들은 (6)의 스튜던트 $t$ 밀도 생성 함수를 채택하면 직접적으로 $\it EST$ 분포로 축소된다. 

이제 본 논문의 구성은 다음과 같다. **제2장**에서는 $\it EST$ 분포의 확률적 성질을 다룬다. 이에 대한 증명은 부록에 제공된다. 이론적 성질에는 다양한 확률적 표현, 주변 및 조건부 분포, 선형 변환, 모멘트, 그리고 특히 Mardia의 다변량 왜도 및 첨도 측정이 포함된다. **제3장**에서는 $\it EST$ 분포의 통계적 성질을 연구한다. 여기에는 우도 추론, 프로파일 로그 우도의 거동, 스코어 벡터, 피셔 정보 행렬이 포함된다. 특히, 단변량 $\it EST$ 분포 (따라서 $\it ST$ 분포도)의 피셔 정보 행렬은 왜도가 0으로 설정되었을 때 비특이적(non-singular)임을 보이는데, 이는 $\it ESN$ 분포 (따라서 $\it SN$ 분포도)와 다른 점이다. 마지막으로, 조건부 $\it EST$ 분포의 수치적 응용을 기밀 데이터 변형(confidential data perturbation) 맥락에서 제시한다.

### Proposition 6 (Moments of univariate ESN)

$Z \sim {\it ESN}(0, 1, \lambda, \tau)$라고 하자. $N_{(-a, \infty)} \overset{d}{=} (N_0 \mid N_0 + a > 0)$로 정의하며, 여기서 $N_0 \sim \mathcal{N}(0, 1)$이다. 이때, $k = 1, 2, 3, 4$에 대한 모멘트 $\mu_k = \mathbb{E}(Z^k)$는 다음과 같다:

- $\mu_1 = \delta \, \mathbb{E}(N_{(-\bar{\tau}, \infty)})$
- $\mu_2 = (1 - \delta^2) + \delta^2 \, \mathbb{E}(N^2_{(-\bar{\tau}, \infty)})$
- $\mu_3 = 3\delta(1 - \delta^2) \, \mathbb{E}(N_{(-\bar{\tau}, \infty)}) + \delta^3 \, \mathbb{E}(N^3_{(-\bar{\tau}, \infty)})$
- $\mu_4 = 3(1 - \delta^2)^2 + 6\delta^2(1 - \delta^2) \, \mathbb{E}(N^2_{(-\bar{\tau}, \infty)}) + \delta^4 \, \mathbb{E}(N^4_{(-\bar{\tau}, \infty)})$

여기서 

- $\mathbb{E}(N_{(-a, \infty)}) = \zeta_1(a) \equiv \dfrac{\phi_1(a)}{\Phi_1(a)}$
- $\mathbb{E}(N^2_{(-a, \infty)}) = 1 - a \, \zeta_1(a)$
- $\mathbb{E}(N^3_{(-a, \infty)}) = (2 + a^2) \, \zeta_1(a)$
- $\mathbb{E}(N^4_{(-a, \infty)}) = 3 - (3a + a^3) \, \zeta_1(a)$

- $\phi_1(a)$: 표준 정규분포의 확률밀도함수(PDF)  
- $\Phi_1(a)$: 표준 정규분포의 누적분포함수(CDF)  
- $\delta = \dfrac{\lambda}{\sqrt{1 + \lambda^2}}$: 비대칭성(skewness) 계수  
- $\bar{\tau} = \dfrac{\tau}{\sqrt{1 + \lambda^2}}$: 정규화된 확장 매개변수 

### Proposition 7 (Moments of univariate EST)

Let $Z \sim {\it EST}(0, 1, \lambda, \nu, \tau)$. Define the truncated $t$ distribution: $T_{(-a, \infty)} \overset{d}{=} (T_0 \mid T_0 + a > 0)$, where $T_0 \sim t(0, 1, \nu)$. Then, for $k = 1, 2, 3, 4$, the moments $\mu_k = \mathbb{E}(Z^k)$ are given by:

- $\mu_1 = \delta \, \mathbb{E}(T_{(-\bar{\tau}, \infty)}), \quad \nu > 1$
- $\mu_2 = \dfrac{\nu}{\nu - 1}(1 - \delta^2)\left(1 + \dfrac{1}{\nu} \mathbb{E}(T^2_{(-\bar{\tau}, \infty)}) \right) + \delta^2 \, \mathbb{E}(T^2_{(-\bar{\tau}, \infty)}), \quad \nu > 2$
- $\mu_3 = \dfrac{3\nu}{\nu - 1} \delta(1 - \delta^2) \left( \mathbb{E}(T_{(-\bar{\tau}, \infty)}) + \dfrac{1}{\nu} \mathbb{E}(T^3_{(-\bar{\tau}, \infty)}) \right) + \delta^3 \, \mathbb{E}(T^3_{(-\bar{\tau}, \infty)}), \quad \nu > 3$
- $\mu_4 = \dfrac{3\nu^2}{(\nu - 1)(\nu - 3)}(1 - \delta^2)^2 \left(1 + \dfrac{2}{\nu} \mathbb{E}(T^2_{(-\bar{\tau}, \infty)}) + \dfrac{1}{\nu^2} \mathbb{E}(T^4_{(-\bar{\tau}, \infty)}) \right)$  
  $\quad \quad \quad + \dfrac{6\nu}{\nu - 1} \delta^2(1 - \delta^2) \left( \mathbb{E}(T^2_{(-\bar{\tau}, \infty)}) + \dfrac{1}{\nu} \mathbb{E}(T^4_{(-\bar{\tau}, \infty)}) \right) + \delta^4 \, \mathbb{E}(T^4_{(-\bar{\tau}, \infty)}), \quad \nu > 4$

여기에서 

- $\mathbb{E}(T_{(-\bar{\tau}, \infty)}) = \dfrac{\nu}{\nu - 1} \left(1 + \dfrac{\bar{\tau}^2}{\nu} \right) t(\bar{\tau}; \nu) / T(\bar{\tau}; \nu), \quad \nu > 1$
- $\mathbb{E}(T^2_{(-\bar{\tau}, \infty)}) = \dfrac{\nu}{\nu - 2} \dfrac{T(\bar{\tau}_2; \nu - 2)}{T(\bar{\tau}; \nu)} - \bar{\tau} \, \mathbb{E}(T_{(-\bar{\tau}, \infty)}), \quad \nu > 2$
- $\mathbb{E}(T^3_{(-\bar{\tau}, \infty)}) = \dfrac{2\nu^2}{(\nu - 1)(\nu - 3)} \left(1 + \dfrac{\bar{\tau}^2}{\nu} \right)^2 t(\bar{\tau}; \nu) / T(\bar{\tau}; \nu) + \bar{\tau}^2 \, \mathbb{E}(T_{(-\bar{\tau}, \infty)}), \quad \nu > 3$
- $\mathbb{E}(T^4_{(-\bar{\tau}, \infty)}) = \dfrac{3\nu^2}{(\nu - 2)(\nu - 4)} \dfrac{T(\bar{\tau}_4; \nu - 4)}{T(\bar{\tau}; \nu)} - \dfrac{3}{2} \bar{\tau} \, \mathbb{E}(T^3_{(-\bar{\tau}, \infty)}) + \dfrac{1}{2} \bar{\tau}^3 \, \mathbb{E}(T_{(-\bar{\tau}, \infty)}), \quad \nu > 4$

- $\delta = \dfrac{\lambda}{\sqrt{1 + \lambda^2}}$
- $\bar{\tau} = \dfrac{\tau}{\sqrt{1 + \lambda^2}}$
- $\bar{\tau}_r = \sqrt{\dfrac{\nu + r}{\nu}} \bar{\tau}, \quad \nu + r > 0$
- $t(\cdot; \nu)$: univariate $t$ density with $\nu$ degrees of freedom  
- $T(\cdot; \nu)$: univariate $t$ CDF with $\nu$ degrees of freedom

```python
import numpy as np
from scipy.stats import t

def delta_lambda(lambda_):
    return lambda_ / np.sqrt(1 + lambda_**2)

def tau_bar(tau, lambda_):
    return tau / np.sqrt(1 + lambda_**2)

def t_pdf(x, nu):
    return t.pdf(x, df=nu)

def T_cdf(x, nu):
    return t.cdf(x, df=nu)

def E_T1(tau_bar, nu):
    return (nu / (nu - 1)) * (1 + tau_bar**2 / nu) * t_pdf(tau_bar, nu) / T_cdf(tau_bar, nu)

def E_T2(tau_bar, nu):
    tau_bar_2 = np.sqrt((nu + 2) / nu) * tau_bar
    return (nu / (nu - 2)) * T_cdf(tau_bar_2, nu - 2) / T_cdf(tau_bar, nu) - tau_bar * E_T1(tau_bar, nu)

def E_T3(tau_bar, nu):
    return (2 * nu**2 / ((nu - 1) * (nu - 3))) * (1 + tau_bar**2 / nu)**2 * t_pdf(tau_bar, nu) / T_cdf(tau_bar, nu) + tau_bar**2 * E_T1(tau_bar, nu)

def E_T4(tau_bar, nu):
    tau_bar_4 = np.sqrt((nu + 4) / nu) * tau_bar
    term1 = (3 * nu**2 / ((nu - 2) * (nu - 4))) * T_cdf(tau_bar_4, nu - 4) / T_cdf(tau_bar, nu)
    term2 = (3 / 2) * tau_bar * E_T3(tau_bar, nu)
    term3 = (1 / 2) * tau_bar**3 * E_T1(tau_bar, nu)
    return term1 - term2 + term3

def est_moments(lambda_, nu, tau):
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)

    mu1 = delta * E_T1(tau_b, nu)
    mu2 = (nu / (nu - 1)) * (1 - delta**2) * (1 + E_T2(tau_b, nu) / nu) + delta**2 * E_T2(tau_b, nu)
    mu3 = (3 * nu / (nu - 1)) * delta * (1 - delta**2) * (E_T1(tau_b, nu) + E_T3(tau_b, nu) / nu) + delta**3 * E_T3(tau_b, nu)
    mu4 = (3 * nu**2 / ((nu - 1) * (nu - 3))) * (1 - delta**2)**2 * \
          (1 + 2 * E_T2(tau_b, nu) / nu + E_T4(tau_b, nu) / nu**2) + \
          (6 * nu / (nu - 1)) * delta**2 * (1 - delta**2) * \
          (E_T2(tau_b, nu) + E_T4(tau_b, nu) / nu) + \
          delta**4 * E_T4(tau_b, nu)

    variance = mu2 - mu1**2
    skewness = (mu3 - 3 * mu1 * variance - mu1**3) / variance**1.5
    kurtosis = (mu4 - 4 * mu3 * mu1 + 6 * mu2 * mu1**2 - 3 * mu1**4) / variance**2

    return {
        "mean": mu1,
        "variance": variance,
        "skewness": skewness,
        "kurtosis": kurtosis
    }

def est_mean(lambda_, nu, tau):
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)

    mu1 = delta * E_T1(tau_b, nu)
    
    return {mu1}

def est_variance(lambda_, nu, tau):
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)

    mu1 = delta * E_T1(tau_b, nu)
    mu2 = (nu / (nu - 1)) * (1 - delta**2) * (1 + E_T2(tau_b, nu) / nu) + delta**2 * E_T2(tau_b, nu)

    variance = mu2 - mu1**2

    return {variance}

def est_skewness(lambda_, nu, tau):
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)

    mu1 = delta * E_T1(tau_b, nu)
    mu2 = (nu / (nu - 1)) * (1 - delta**2) * (1 + E_T2(tau_b, nu) / nu) + delta**2 * E_T2(tau_b, nu)
    mu3 = (3 * nu / (nu - 1)) * delta * (1 - delta**2) * (E_T1(tau_b, nu) + E_T3(tau_b, nu) / nu) + delta**3 * E_T3(tau_b, nu)
    mu4 = (3 * nu**2 / ((nu - 1) * (nu - 3))) * (1 - delta**2)**2 * \
          (1 + 2 * E_T2(tau_b, nu) / nu + E_T4(tau_b, nu) / nu**2) + \
          (6 * nu / (nu - 1)) * delta**2 * (1 - delta**2) * \
          (E_T2(tau_b, nu) + E_T4(tau_b, nu) / nu) + \
          delta**4 * E_T4(tau_b, nu)

    variance = mu2 - mu1**2
    skewness = (mu3 - 3 * mu1 * variance - mu1**3) / variance**1.5
    kurtosis = (mu4 - 4 * mu3 * mu1 + 6 * mu2 * mu1**2 - 3 * mu1**4) / variance**2

    return {skewness}

def est_kurtosis(lambda_, nu, tau):
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)

    mu1 = delta * E_T1(tau_b, nu)
    mu2 = (nu / (nu - 1)) * (1 - delta**2) * (1 + E_T2(tau_b, nu) / nu) + delta**2 * E_T2(tau_b, nu)
    mu3 = (3 * nu / (nu - 1)) * delta * (1 - delta**2) * (E_T1(tau_b, nu) + E_T3(tau_b, nu) / nu) + delta**3 * E_T3(tau_b, nu)
    mu4 = (3 * nu**2 / ((nu - 1) * (nu - 3))) * (1 - delta**2)**2 * \
          (1 + 2 * E_T2(tau_b, nu) / nu + E_T4(tau_b, nu) / nu**2) + \
          (6 * nu / (nu - 1)) * delta**2 * (1 - delta**2) * \
          (E_T2(tau_b, nu) + E_T4(tau_b, nu) / nu) + \
          delta**4 * E_T4(tau_b, nu)

    variance = mu2 - mu1**2
    skewness = (mu3 - 3 * mu1 * variance - mu1**3) / variance**1.5
    kurtosis = (mu4 - 4 * mu3 * mu1 + 6 * mu2 * mu1**2 - 3 * mu1**4) / variance**2

    return {kurtosis}
```

### CDF of the EST

$\it EST$ 분포의 누적분포함수(CDF)는 식 (7) 또는 (9)를 이용하여 쉽게 계산할 수 있다. 만약 $Y \sim \mathrm{EST}_p(\xi, \Omega, \lambda, \nu, \tau)$라면, 다음이 성립한다:

$$
P(Y \leq y) = \frac{1}{T_1(\bar{\tau}; \nu)} \, T_{p+1}
\left(
\begin{bmatrix}
z \\
\bar{\tau}
\end{bmatrix}
;
\begin{bmatrix}
\bar{\Omega} & -\delta \\
-\delta^\top & 1
\end{bmatrix},
\nu + 1
\right),
$$

여기서:

- $z = \Omega^{-1}(y - \xi)$  
- $\bar{\tau} = \dfrac{\tau}{\sqrt{1 + \lambda^\top \lambda}}$  
- $\delta = \dfrac{\lambda}{\sqrt{1 + \lambda^\top \lambda}}$  
- $\bar{\Omega} = \Omega$ (혹은 관련된 변환에 따라 다를 수 있음)  
- $T_k(\cdot; \Sigma, \nu)$는 자유도 $\nu$와 공분산 $\Sigma$를 갖는 $k$-변량 $t$ 분포의 누적분포함수

### 단변량 EST 분포의 누적분포함수 (CDF)

만약 $ Y \sim \mathrm{EST}_1(\xi, \omega^2, \lambda, \nu, \tau) $라면, 그 누적분포함수는 다음과 같다:

$$
P(Y \leq y) = \frac{1}{T_1(\bar{\tau}; \nu)} \, T_2
\left(
\begin{bmatrix}
z \\
\bar{\tau}
\end{bmatrix}
;
\begin{bmatrix}
1 & -\delta \\
-\delta & 1
\end{bmatrix},
\nu + 1
\right),
$$

여기서:

- $ z = \dfrac{y - \xi}{\omega} $
- $ \bar{\tau} = \dfrac{\tau}{\sqrt{1 + \lambda^2}} $
- $ \delta = \dfrac{\lambda}{\sqrt{1 + \lambda^2}} $
- $ T_2(\cdot; \Sigma, \nu) $는 자유도 $ \nu $와 공분산 행렬 $ \Sigma $를 갖는 2변량 Student's $ t $ 누적분포함수

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t, multivariate_t

# EST1 CDF function
def est1_cdf(y, xi, omega, lam, nu, tau):
    delta = lam / np.sqrt(1 + lam**2)
    tau_bar = tau / np.sqrt(1 + lam**2)
    z = (y - xi) / omega

    # Bivariate t distribution setup
    mean = np.zeros(2)
    cov = np.array([[1, -delta], [-delta, 1]])

    # Denominator: marginal t CDF
    T1_tau_bar = t.cdf(tau_bar, df=nu)

    # Numerator: bivariate t CDF
    x_vec = np.array([z, tau_bar])
    numerator = multivariate_t.cdf(x_vec, df=nu+1, loc=mean, shape=cov)

    return numerator / T1_tau_bar

# Parameters
xi = 0
omega = 1
lam = 2
nu = 3
tau = 0

# y range
y_vals = np.linspace(-4, 4, 500)
cdf_vals = [est1_cdf(y, xi, omega, lam, nu, tau) for y in y_vals]

# Plot
plt.figure(figsize=(8, 5))
plt.plot(y_vals, cdf_vals, label=f'EST1({xi}, {omega**2}, {lam}, {nu}, {tau})')
plt.title('CDF of EST1 Distribution')
plt.xlabel('y')
plt.ylabel('P(Y ≤ y)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
