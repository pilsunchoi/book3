# 확장 비대칭 t 분포

출처: Arellano-Valle, R.B., Genton, M.G. Multivariate extended skew-t distributions and related families. METRON 68, 201–234 (2010). https://doi.org/10.1007/BF03263536

### 정의 1 (확장된 비대칭 t-분포)  
확률변수 $X$가 확장된 비대칭 $t$ 분포(Extended Skew-t, $\it EST$)를 따른다고 할 때, 이를 $X \sim {\it EST}(\xi, \omega, \lambda, \nu, \tau)$로 표기하며, 그 밀도 함수는 다음과 같이 정의된다.  

$$
\frac{1}{T\left({\tau}/{\sqrt{1 + \lambda^2}}; \nu \right)} \ t(x; \xi, \omega, \nu) \ T \left\{ \left(\lambda z + \tau \right) \left(\frac{\nu + 1}{\nu + z^2}\right)^{1/2} ; \nu + 1 \right\}, \tag{4}
$$

여기서 $z = (x - \xi)/\omega$이고, $\omega > 0$, $\nu > 0$이다. $T(x; \nu)$는 자유도 $\nu > 0$를 갖는 일변량 표준 스튜던트 $t$ 누적분포함수(CDF)를 나타낸다. 또한, $t(x; \xi, \omega, \nu)$는 스튜던트 $t$ 분포의 밀도 함수로서 다음과 같다. 

$$
t(x; \xi, \omega, \nu) = \frac{\Gamma((\nu + 1)/2)}{(\pi \nu)^{1/2} \omega \Gamma(\nu/2)}
\left( 1 + \frac{z^2}{\nu} \right)^{-(\nu + 1)/2}
$$

여기서 $z = (x - \xi)/\omega$이고, $\omega > 0$, $\nu > 0$이다. 

### 로그가능도 함수(Log-likelihood function)

$\it EST$ 분포의 로그가능도 함수는 다음과 같이 주어진다:

$$
\ln L = \sum_{i=1}^n \ln f(x_i) = \sum_{i=1}^n \Bigg[ - \ln T\left(\frac{\tau}{\sqrt{1 + \lambda^2}}; \nu \right) 
+ \ln t(x_i; \xi, \omega, \nu)
+ \ln T \left\{ \left(\lambda z_i + \tau \right) \left(\frac{\nu + 1}{\nu + z_i^2}\right)^{1/2} ; \nu + 1 \right\}
\Bigg]
$$
여기서
$$
\ln t(x_i; \xi, \omega, \nu) = \ln \Gamma \left(\frac{\nu + 1}{2}\right) - \frac{1}{2}\ln(\pi \nu) - \ln\omega - \ln\Gamma\left(\frac{\nu}{2}\right) - \frac{\nu + 1}{2}\ln\left( 1 + \frac{z_i^2}{\nu} \right)
$$
여기서 $z_i = (x_i - \xi)/\omega$이다.

### 다변량 $\it EST$ 분포

연속적인 $p$-차원 확률 벡터 $X$가 다변량 확장 비대칭 $t$ 분포(Extended Skew-t, EST)를 따른다고 할 때, 이를 $X \sim EST_p(\xi, \Omega, \lambda, \nu, \tau)$로 표기하며, 그 **밀도 함수**는 $x \in \mathbb{R}^p$에서 다음과 같이 정의된다.  

$$
\frac{1}{T_1\left({\tau}/{\sqrt{1 + \lambda' \bar{\Omega} \lambda}}; \nu \right)}
t_p(x; \xi, \Omega, \nu) \ T_1 \left\{ \left(\lambda' z + \tau \right) \left(\frac{\nu + p}{\nu + Q(z)}\right)^{1/2} ; \nu + p \right\}, \tag{1}
$$

여기서 $z = \omega^{-1} (x - \xi)$, $Q(z) = z' \bar{\Omega}^{-1} z$, $\lambda \in \mathbb{R}^p$는 형태(shape) 파라미터, $\tau \in \mathbb{R}$는 확장(extension) 파라미터이다. 또한, $t_p(x; \xi, \Omega, \nu)$는 $p$-차원 Student’s $t$ 분포의 밀도 함수로서 다음과 같다. 

$$
t_p(x; \xi, \Omega, \nu) = \frac{\Gamma((\nu + p)/2)}{|\Omega|^{1/2} (\pi \nu)^{p/2} \Gamma(\nu/2)}
\left( 1 + \frac{Q(z)}{\nu} \right)^{-(\nu + p)/2}
$$

여기서 $\xi \in \mathbb{R}^p$는 위치(location) 파라미터, $\Omega$는 양의 정부호 $p \times p$ 분산 행렬, $\omega = \text{diag}(\Omega)^{1/2}$는 $p \times p$ 스케일 행렬, $\bar{\Omega} = \omega^{-1} \Omega \omega^{-1}$는 상관 행렬, $\nu > 0$는 자유도(degrees of freedom)이다. 마지막으로, $T_1(x; \nu)$는 자유도 $\nu > 0$를 갖는 일변량 표준 Student’s $t$ 누적분포함수(CDF)를 나타낸다.

### 이변량 $\it EST$ 분포

두 연속 확률변수 $X = (X_1, X_2)'$가 이변량 확장 비대칭 $t$ 분포(Extended Skew-$t$ distribution, $\it EST$)를 따른다면, 이를  
$$
X \sim EST_2(\xi, \Omega, \lambda, \nu, \tau)
$$  
라고 쓰며, 그 밀도 함수는 $x = (x_1, x_2)' \in \mathbb{R}^2$에서 다음과 같이 정의된다:

$$
\frac{1}{T_1\left(\dfrac{\tau}{\sqrt{1 + \lambda' \bar{\Omega} \lambda}}; \nu \right)}
t_2(z; \xi, \Omega, \nu) \cdot T_1 \left\{ \left(\lambda' z + \tau \right) \left(\frac{\nu + 2}{\nu + Q(z)}\right)^{1/2} ; \nu + 2 \right\}, \tag{1}
$$

여기에서:

- $z = \omega^{-1}(x - \xi)$는 표준화된 변수 벡터이고,  
- $Q(z) = z' \bar{\Omega}^{-1} z$는 마할라노비스 제곱 거리,  
- $T_1(\cdot; \nu)$는 자유도 $\nu$를 갖는 일변량 $t$ 누적분포함수(CDF)이다.
- $\xi = (\xi_1, \xi_2)' \in \mathbb{R}^2$는 위치(location) 파라미터,  
- $\Omega$는 양의 정부호 $2 \times 2$ 분산 행렬,  
- $\omega = \operatorname{diag}(\Omega)^{1/2}$는 스케일 행렬 (즉, $\omega = \operatorname{diag}(\sqrt{\Omega_{11}}, \sqrt{\Omega_{22}})$),  
- $\bar{\Omega} = \omega^{-1} \Omega \omega^{-1}$는 상관 행렬,  
- $\nu > 0$는 자유도(degrees of freedom)이다.

#### $t_2(x; \xi, \Omega, \nu)$의 정의

이변량 Student's $t$ 분포의 밀도 함수는 다음과 같다:

$$
t_2(x; \xi, \Omega, \nu) = \frac{1}{2 \pi |\Omega|^{1/2}} \left( 1 + \frac{Q(z)}{\nu} \right)^{-(\nu + 2)/2}
$$

여기서 $|\Omega|$는 $\Omega$의 행렬식이다.

### 이변량 $\it EST$ 분포 전개

두 연속 확률변수 $X = (X_1, X_2)'$가 이변량 확장 비대칭 $t$ 분포(Extended Skew-$t$ distribution, $\it EST$)를 따른다면, 이를  
$$
X \sim EST_2(\xi, \Omega, \lambda, \nu, \tau)
$$  
$$
X \sim EST_2(\xi_1, \omega_1, \lambda_1, \xi_2, \omega_2, \lambda_2, \nu, \tau, \rho)
$$  
라고 쓰며, 그 밀도 함수는 $x = (x_1, x_2)' \in \mathbb{R}^2$에서 다음과 같이 정의된다:

$$
\frac{1}{T_1\left(\dfrac{\tau}{\sqrt{1 + \lambda' R \lambda}}; \nu \right)}
t_2(z; \xi, \Omega, \nu) \cdot T_1 \left\{ \left(\lambda' z + \tau \right) \left(\frac{\nu + 2}{\nu + Q(z)}\right)^{1/2} ; \nu + 2 \right\}, \tag{1}
$$

여기에서:

- $z_1 = \frac{x_1 - \xi_1}{\omega_1}$, $z_2 = \frac{x_2 - \xi_2}{\omega_2}$,
- $Q(z) = z' R^{-1} z = \frac{1}{(1 - \rho^2)} (z_1^2 - 2\rho z_1 z_2 + z_2^2) $,
- $t_2(x; \xi, \Omega, \nu) = \frac{1}{2\pi \omega_1 \omega_2 \sqrt{1 - \rho^2}} \left( 1 + \frac{1}{\nu} Q(z) \right)^{-(\nu + 2)/2}$,
- $T_1(\cdot; \nu)$: 자유도 $\nu$를 갖는 일변량 $t$ 누적분포함수(CDF),
- $\lambda' R \lambda = \lambda_1^2 + 2\rho \lambda_1 \lambda_2 + \lambda_2^2$
- $\lambda' z = \lambda_1 z_1 + \lambda_2 z_2 $

또한

- $R = \begin{pmatrix} 1 & \rho \\ \rho & 1 \end{pmatrix}$는 상관 행렬,
- $\Omega = \omega' R \omega = \begin{pmatrix} \omega_1^2 & \rho \omega_1 \omega_2 \\ \rho \omega_1 \omega_2 & \omega_2^2 \end{pmatrix}$는 양의 정부호 $2 \times 2$ 분산 행렬

```python
    # Q(z)
    Q = (z1**2 - 2 * rho * z1 * z2 + z2**2) / (1 - rho**2)

    # 공통 상수
    const = (
        gammaln((nu + 2) / 2)
        - gammaln(nu / 2)
        - np.log(2 * np.pi)
        - np.log(omega1)
        - np.log(omega2)
        - 0.5 * np.log(1 - rho**2)
    )
```

다음은 주어진 이변량 확장 비대칭 $t$ 분포 $\mathrm{EST}_2$의 **로그가능도 함수**를 마크다운 형식으로 표현한 것이다. 이 표현은 하나의 관측치 $x = (x_1, x_2)'$에 대한 로그가능도를 정의하며, 여러 관측치가 있을 경우 이 표현을 합산하면 된다.

---

### 로그가능도 함수 (Log-Likelihood Function) of $\it EST_2$

주어진 확률변수  
$$
X \sim EST_2(\xi_1, \omega_1, \lambda_1, \xi_2, \omega_2, \lambda_2, \nu, \tau, \rho)
$$  
에 대해, 관측값 $x = (x_1, x_2)'$에 대한 **로그가능도 함수**는 다음과 같다:

$$
\log f(x_1, x_2) =
- \log T_1\left( \frac{\tau}{\sqrt{1 + \lambda' R \lambda}}; \nu \right)
+ \log \left[ t_2(z; \xi, \Omega, \nu) \right]
+ \log T_1\left( \left( \lambda' z + \tau \right) \left( \frac{\nu + 2}{\nu + Q(z)} \right)^{1/2}; \nu + 2 \right)
$$

#### 각 항목의 정의:

- **표준화된 변수 $z_1, z_2$**:
  $$
  z_1 = \frac{x_1 - \xi_1}{\omega_1}, \quad z_2 = \frac{x_2 - \xi_2}{\omega_2}
  $$

- **$Q(z)$**:
  $$
  Q(z) = \frac{1}{1 - \rho^2} (z_1^2 - 2\rho z_1 z_2 + z_2^2)
  $$

- **밀도항 $t_2$의 로그**:
  $$
  \log t_2(z; \xi, \Omega, \nu) = -\log(2\pi \omega_1 \omega_2 \sqrt{1 - \rho^2})
  - \frac{\nu + 2}{2} \log \left( 1 + \frac{Q(z)}{\nu} \right)
  $$

- **형태항 $\lambda' z$**:
  $$
  \lambda' z = \lambda_1 z_1 + \lambda_2 z_2 = \lambda_1 \left( \frac{x_1 - \xi_1}{\omega_1} \right) + \lambda_2 \left( \frac{x_2 - \xi_2}{\omega_2} \right)
  $$

- **상수항 $\lambda' R \lambda$**:
  $$
  \lambda' R \lambda = \lambda_1^2 + 2\rho \lambda_1 \lambda_2 + \lambda_2^2
  $$

- **확장 비대칭항 $T_1$의 인수**:
  $$
  h(x) = \left( \lambda' z + \tau \right) \left( \frac{\nu + 2}{\nu + Q(z)} \right)^{1/2}
  $$

- **최종 로그가능도**는 다음과 같이 정리된다:
  $$
  \log f(x_1, x_2) =
  - \log T_1\left( \frac{\tau}{\sqrt{1 + \lambda_1^2 + 2\rho \lambda_1 \lambda_2 + \lambda_2^2}}; \nu \right)
  - \log(2\pi \omega_1 \omega_2 \sqrt{1 - \rho^2})
  - \frac{\nu + 2}{2} \log \left( 1 + \frac{Q(z)}{\nu} \right)
  + \log T_1\left( h(x); \nu + 2 \right)
  $$

---

필요하다면 이 식을 수치 계산용 함수로도 만들어줄 수 있어.

```python
    # log t 밀도
    log_t_density = const - ((nu + 2) / 2) * np.log(1 + Q / nu)

    # 스큐 관련 항들
    lam_dot_z = lam1 * z1 + lam2 * z2
    bar_tau = tau / np.sqrt(1 + lam1**2 + 2 * rho * lam1 * lam2 + lam2**2)
    skew_arg = (lam_dot_z + tau) * np.sqrt((nu + 2) / (nu + Q))

    # 누적분포함수 항
    T_bar_tau = t.cdf(bar_tau, df=nu)
    T_skew_part = t.cdf(skew_arg, df=nu + 2)

    # 로그우도 계산
    ln_pdf_vals = -np.log(T_bar_tau) + log_t_density + np.log(T_skew_part)
```

## 참고: 표준 이변량 $t$ 분포의 밀도 함수

#### 표준 이변량 Student's $t$ 분포 (Standard Bivariate $t$-distribution)

자유도 $\nu$를 갖는 **표준 이변량 $t$ 분포** (즉, 평균이 0이고, 단위 스케일을 갖는 경우)는 다음과 같은 밀도 함수로 정의된다:

$$
f(t_1, t_2) = \frac{|\mathbf{A}|^{1/2}}{2\pi} \left(1 + \sum_{i,j=1}^{2} \frac{A_{ij} t_i t_j}{\nu} \right)^{-(\nu + 2)/2}
$$

여기서:

- $\mathbf{A}$는 대칭이고 양의 정부호인 $2 \times 2$ 행렬,
- $\nu$는 자유도(degrees of freedom),
- $t_1, t_2$는 관측값이다.

또한, $A_{ij}$는 $\mathbf{A}$의 $(i, j)$ 원소이다.

참고로 다음 항등식이 성립한다:

$$
\frac{\Gamma\left(\frac{\nu + 2}{2}\right)}{\nu \, \Gamma\left(\frac{\nu}{2}\right)} = \frac{1}{2}
$$

따라서 $\mathbf{A} = \nu \, \Sigma^{-1}$로 놓으면, $f(t_1, t_2)$는 이변량 $t$ 분포의 정규화된 형태를 갖는다.

필요하다면 $\mathbf{A}$를 공분산 행렬 $\Sigma$의 역행렬과 자유도를 이용해 구성하는 방식까지 포함해서 확장할 수도 있어.

## 참고: 공분산 행렬 $\Sigma$의 역행렬과 자유도를 이용해 표현

#### 표준 이변량 Student's $t$ 분포 (Standard Bivariate $t$-distribution)

자유도 $\nu$를 갖는 **표준 이변량 $t$ 분포**의 밀도 함수는 다음과 같이 정의된다:

$$
f(t_1, t_2) = \frac{1}{2\pi \, |\Sigma|^{1/2}} \left( 1 + \frac{1}{\nu} \mathbf{t}' \Sigma^{-1} \mathbf{t} \right)^{-(\nu + 2)/2}
$$

여기서,

- $\mathbf{t} = \begin{pmatrix} t_1 \\ t_2 \end{pmatrix}$는 2차원 벡터,
- $\Sigma$는 $2 \times 2$ 양의 정부호 공분산 행렬,
- $\nu > 0$는 자유도,
- $|\Sigma|$는 $\Sigma$의 행렬식이다.

이 정의는 다변량 Student's $t$ 분포의 일반적인 형식을 따르며, $p = 2$인 경우에 해당한다.

#### $\mathbf{A}$ 행렬과의 관계

이 분포를 다음과 같이 쓸 수도 있다:

$$
f(t_1, t_2) = \frac{|\mathbf{A}|^{1/2}}{2\pi} \left( 1 + \sum_{i,j=1}^{2} \frac{A_{ij} t_i t_j}{\nu} \right)^{-(\nu + 2)/2}
$$

여기서 $\mathbf{A} = \nu \, \Sigma^{-1}$로 정의하면, 위 두 표현은 서로 동등하다. 따라서:

- $\Sigma^{-1} = \mathbf{A} / \nu$
- $|\mathbf{A}| = \nu^2 |\Sigma^{-1}| = \nu^2 / |\Sigma|$

이 관계를 활용하면 표준 이변량 $t$ 분포의 밀도 함수를 다양한 방식으로 쓸 수 있다.

필요하다면 $\Sigma$의 구체적인 형태 (예: 공분산 $\rho$를 포함하는 형태)로도 확장해서 표현해 줄 수 있어.

## 참고: 전개된 형태

#### 표준 이변량 Student’s $t$ 분포 (Standard Bivariate $t$-distribution)

자유도 $\nu$를 갖고 공분산 $\rho \in (-1, 1)$인 **표준 이변량 Student's $t$ 분포**에서, 벡터 $\mathbf{t} = \begin{pmatrix} t_1 \\ t_2 \end{pmatrix}$에 대한 밀도 함수는 다음과 같다:

$$
f(t_1, t_2) = \frac{1}{2 \pi \nu\ |\Sigma|^{1/2}} \left( 1 + \frac{1}{\nu} \mathbf{t}' \Sigma^{-1} \mathbf{t} \right)^{-\frac{\nu + 2}{2}}
$$

여기서,

- 공분산 행렬 $\Sigma$는 다음과 같이 정의된다:

$$
\Sigma = \begin{pmatrix} 1 & \rho \\ \rho & 1 \end{pmatrix}
$$

- 그 역행렬은 다음과 같다:

$$
\Sigma^{-1} = \frac{1}{1 - \rho^2}
\begin{pmatrix}
1 & -\rho \\
-\rho & 1
\end{pmatrix}
$$

- 그리고 행렬식은:

$$
|\Sigma| = 1 - \rho^2
$$

따라서 밀도 함수는 다음과 같이 정리된다:

$$
f(t_1, t_2) = \frac{1}{2\pi \sqrt{1 - \rho^2}} \left[ 1 + \frac{1}{\nu (1 - \rho^2)} (t_1^2 - 2\rho t_1 t_2 + t_2^2) \right]^{-\frac{\nu + 2}{2}}
$$


#### 요약

- $t_1, t_2$: 주변 변수 (표준 정규화됨)
- $\nu$: 자유도
- $\rho$: 두 변수 간의 공분산
- 위 식은 $\rho = 0$일 때, 두 $t$ 분포가 독립일 때의 이변량 $t$ 분포로 환원됨

필요하면 이 밀도를 이용해 샘플링이나 적분 같은 것도 가능해.

## 참고: $z$ 대신 $x$로 다시 표현 

위에서 정의한  
$$ z = \omega^{-1}(x - \xi) \quad \text{즉 } z_i = \frac{x_i - \xi_i}{\omega_i} $$  
를 $z$ 대신 $x$로 다시 표현하면, 표준 이변량 Student’s $t$ 분포의 밀도 함수는 다음과 같이 다시 쓸 수 있어:

#### 표준 이변량 Student’s $t$ 분포 (위치 및 스케일 포함 표현)

자유도 $\nu$, 위치 벡터 $\xi = (\xi_1, \xi_2)^T$, 스케일 벡터 $\omega = (\omega_1, \omega_2)^T$, 공분산 계수 $\rho \in (-1, 1)$를 갖는 이변량 $t$ 분포의 밀도 함수는 다음과 같다:

$$
f(x_1, x_2) = \frac{1}{2\pi \omega_1 \omega_2 \sqrt{1 - \rho^2}} \left[ 1 + \frac{1}{\nu (1 - \rho^2)} Q(x) \right]^{-\frac{\nu + 2}{2}}
$$

여기서 $Q(x)$는 $x$의 정규화된 이차형식으로 다음과 같이 정의된다:

$$
Q(x) = \left( \frac{x_1 - \xi_1}{\omega_1} \right)^2 - 2\rho \left( \frac{x_1 - \xi_1}{\omega_1} \right)\left( \frac{x_2 - \xi_2}{\omega_2} \right) + \left( \frac{x_2 - \xi_2}{\omega_2} \right)^2
$$

#### 요약

- 이 식은 원래 변수 $x = (x_1, x_2)^T$에 대해 직접 계산할 수 있도록 표현된 형태야.
- $\omega_1, \omega_2$는 각각 $x_1, x_2$의 표준편차
- $\xi_1, \xi_2$는 각각의 평균
- $\rho$는 상관계수

필요하면 이걸 일반 $p$-차원으로도 확장 가능해.

## 참고: 아래 두 식은 같은 밀도 함수를 서로 다른 표현 방식으로 쓴 것임

#### 1. 너가 준 표현 (행렬식과 $z$를 이용):

$$
t_2(x; \xi, \Omega, \nu) = \frac{1}{2 \pi |\Omega|^{1/2}} \left( 1 + \frac{Q(z)}{\nu} \right)^{-\frac{\nu + 2}{2}}, \quad \text{where } z = \omega^{-1}(x - \xi)
$$

여기서  
- $\Omega = \omega R \omega$는 공분산 행렬  
- $Q(z) = z' R^{-1} z$ (즉, 정규화된 이차형식)

#### 2. 아까 표현한 방식 ($x$를 기준으로 전개):

$$
f(x_1, x_2) = \frac{1}{2\pi \omega_1 \omega_2 \sqrt{1 - \rho^2}} \left[ 1 + \frac{1}{\nu (1 - \rho^2)} Q(x) \right]^{-\frac{\nu + 2}{2}}
$$

여기서  
- $\Omega = \begin{pmatrix} \omega_1^2 & \rho \omega_1 \omega_2 \\ \rho \omega_1 \omega_2 & \omega_2^2 \end{pmatrix}$  
- $|\Omega| = \omega_1^2 \omega_2^2 (1 - \rho^2)$  
- $Q(x) = \left( \frac{x_1 - \xi_1}{\omega_1} \right)^2 - 2\rho \left( \frac{x_1 - \xi_1}{\omega_1} \right)\left( \frac{x_2 - \xi_2}{\omega_2} \right) + \left( \frac{x_2 - \xi_2}{\omega_2} \right)^2$

#### 결론

결국 두 식은,

- $z = \omega^{-1}(x - \xi)$를 쓰느냐  
- 아니면 이를 직접 전개해서 $x$로 표현하느냐  

의 차이일 뿐이고, 같은 분포를 나타낸다. 따라서 **동일한 식** 맞아.

## 참고: 전개된 형태로 표현

좋아. 2차원($p = 2$)일 때, $\bar{\Omega} = \omega^{-1} \Omega \omega^{-1}$을 **행렬곱이 아닌 전개된 원소 형태**로 풀어보면 다음과 같아.

#### 주어진 것들:

- $\Omega = \begin{pmatrix}
\omega_1^2 & \rho \omega_1 \omega_2 \\
\rho \omega_1 \omega_2 & \omega_2^2
\end{pmatrix}$  
 (공분산 행렬)

- $\omega = \text{diag}(\omega_1, \omega_2)$  
 (스케일 행렬)

- $\omega^{-1} = \text{diag}(1/\omega_1, 1/\omega_2)$

#### 계산:

$$
\bar{\Omega} = \omega^{-1} \Omega \omega^{-1} =
\begin{pmatrix}
1/\omega_1 & 0 \\
0 & 1/\omega_2
\end{pmatrix}
\begin{pmatrix}
\omega_1^2 & \rho \omega_1 \omega_2 \\
\rho \omega_1 \omega_2 & \omega_2^2
\end{pmatrix}
\begin{pmatrix}
1/\omega_1 & 0 \\
0 & 1/\omega_2
\end{pmatrix}
$$

차례대로 곱하면:

1. 먼저 왼쪽에서 곱하면:


$$
\begin{pmatrix}
\omega_1 & \rho \omega_2 \\
\rho \omega_1 & \omega_2
\end{pmatrix}
$$

2. 다시 오른쪽에서 곱하면:

$$
\bar{\Omega} = 
\begin{pmatrix}
\omega_1 & \rho \omega_2 \\
\rho \omega_1 & \omega_2
\end{pmatrix}
\begin{pmatrix}
1/\omega_1 & 0 \\
0 & 1/\omega_2
\end{pmatrix}
=\begin{pmatrix}
1 & \rho \\
\rho & 1
\end{pmatrix}
$$

#### 결론

따라서 전개된 형태로 쓰면:

$$
\bar{\Omega} =
\begin{pmatrix}
1 & \rho \\
\rho & 1
\end{pmatrix}
$$

즉, $\bar{\Omega}$는 공분산 행렬 $\Omega$에서 스케일을 제거한 **상관 행렬**로서, 단순히 $\rho$만 들어간 대칭행렬이 된다.

## 참고: $\lambda^T \bar{\Omega} \lambda$ 2차원에서 전개된 형태

#### 1. 주어진 것들

- $\lambda = \begin{pmatrix} \lambda_1 \\ \lambda_2 \end{pmatrix}$
- $\bar{\Omega} = \begin{pmatrix} 1 & \rho \\ \rho & 1 \end{pmatrix}$

#### 2. 계산

$$
\lambda^T \bar{\Omega} \lambda = 
\begin{pmatrix} \lambda_1 & \lambda_2 \end{pmatrix}
\begin{pmatrix} 1 & \rho \\ \rho & 1 \end{pmatrix}
\begin{pmatrix} \lambda_1 \\ \lambda_2 \end{pmatrix}
$$

중간 행렬곱 먼저 하면:

$$
\begin{pmatrix} \lambda_1 & \lambda_2 \end{pmatrix}
\begin{pmatrix}
\lambda_1 + \rho \lambda_2 \\
\rho \lambda_1 + \lambda_2
\end{pmatrix}
$$

이걸 계산하면:

$$
\lambda_1 (\lambda_1 + \rho \lambda_2) + \lambda_2 (\rho \lambda_1 + \lambda_2)
= \lambda_1^2 + 2\rho \lambda_1 \lambda_2 + \lambda_2^2
$$

#### 3. 결론

$$
\lambda^T \bar{\Omega} \lambda = \lambda_1^2 + 2\rho \lambda_1 \lambda_2 + \lambda_2^2
$$

즉, 이것도 단순한 2차 정규형으로 정리된다.

## 2차 정규형이란?

"2차 정규형"이란 건 2차 형식(quadratic form)의 특별한 경우로, 다음과 같이 생긴 식을 말해:

$$
x^T A x
$$

여기서  
- $x$는 열 벡터 (예: $x = \begin{pmatrix} x_1 \\ x_2 \end{pmatrix}$)  
- $A$는 대칭 행렬 (예: $\begin{pmatrix} a & b \\ b & c \end{pmatrix}$)

이걸 전개하면 이렇게 돼:

$$
x^T A x = a x_1^2 + 2b x_1 x_2 + c x_2^2
$$

이게 바로 우리가 앞에서 구한 형태랑 똑같아:

$$
\lambda^T \bar{\Omega} \lambda = \lambda_1^2 + 2\rho \lambda_1 \lambda_2 + \lambda_2^2
$$

- 여기서 $a = 1$, $b = \rho$, $c = 1$인 셈이고,
- $A = \bar{\Omega} = \begin{pmatrix} 1 & \rho \\ \rho & 1 \end{pmatrix}$

요약하면, **2차 정규형**이란 건 "벡터의 행렬곱으로 표현되는 2차 다항식"이고, 특히 **대칭 행렬**이 들어간 이 형태가 많이 쓰여. 통계, 최적화, 분포 이론 등에서 아주 자주 나오는 구조야.

```python
import numpy as np
from scipy.stats import t
from scipy.special import gamma

def t_pdf(y, xi, omega, nu):
    z = (y - xi) / omega
    coef = gamma((nu + 1) / 2) / (np.sqrt(np.pi * nu) * omega * gamma(nu / 2))
    density = coef * (1 + z**2 / nu) ** (-(nu + 1) / 2)
    return density
```

```python
import numpy as np
from scipy.stats import t
from scipy.special import gamma

def EST_pdf(y, xi, omega, lam, nu, tau):

    z = (y - xi) / omega
    
    # denominator normalization term
    denom_cdf = t.cdf(tau / np.sqrt(1 + lam**2), df=nu)
    
    # t density
    t_density = t_pdf(y, xi, omega, nu)
    
    # transformed argument for second CDF
    arg = (lam * z + tau) * np.sqrt((nu + 1) / (nu + z**2))
    cdf_term = t.cdf(arg, df=nu + 1)
    
    return (1 / denom_cdf) * t_density * cdf_term
```

### Proposition 6 (Moments of univariate ESN)

$Z \sim {\it ESN}(0, 1, \lambda, \tau)$라고 하자. $N_{(-a, \infty)} \overset{d}{=} (N_0 \mid N_0 + a > 0)$로 정의하며, 여기서 $N_0 \sim \mathcal{N}(0, 1)$이다. 이때, $k = 1, 2, 3, 4$에 대한 모멘트 $\mu_k = \mathbb{E}(Z^k)$는 다음과 같다:

- $\mu_1 = \delta \, \mathbb{E}(N_{(-\bar{\tau}, \infty)})$
- $\mu_2 = (1 - \delta^2) + \delta^2 \, \mathbb{E}(N^2_{(-\bar{\tau}, \infty)})$
- $\mu_3 = 3\delta(1 - \delta^2) \, \mathbb{E}(N_{(-\bar{\tau}, \infty)}) + \delta^3 \, \mathbb{E}(N^3_{(-\bar{\tau}, \infty)})$
- $\mu_4 = 3(1 - \delta^2)^2 + 6\delta^2(1 - \delta^2) \, \mathbb{E}(N^2_{(-\bar{\tau}, \infty)}) + \delta^4 \, \mathbb{E}(N^4_{(-\bar{\tau}, \infty)})$

여기서 

- $\mathbb{E}(N_{(-a, \infty)}) = \zeta_1(a) \equiv \dfrac{\phi_1(a)}{\Phi_1(a)}$
- $\mathbb{E}(N^2_{(-a, \infty)}) = 1 - a \, \zeta_1(a)$
- $\mathbb{E}(N^3_{(-a, \infty)}) = (2 + a^2) \, \zeta_1(a)$
- $\mathbb{E}(N^4_{(-a, \infty)}) = 3 - (3a + a^3) \, \zeta_1(a)$

- $\phi_1(a)$: 표준 정규분포의 확률밀도함수(PDF)  
- $\Phi_1(a)$: 표준 정규분포의 누적분포함수(CDF)  
- $\delta = \dfrac{\lambda}{\sqrt{1 + \lambda^2}}$: 비대칭성(skewness) 계수  
- $\bar{\tau} = \dfrac{\tau}{\sqrt{1 + \lambda^2}}$: 정규화된 확장 매개변수 

### Proposition 7 (Moments of univariate EST)

Let $Z \sim {\it EST}(0, 1, \lambda, \nu, \tau)$. Define the truncated $t$ distribution: $T_{(-a, \infty)} \overset{d}{=} (T_0 \mid T_0 + a > 0)$, where $T_0 \sim t(0, 1, \nu)$. Then, for $k = 1, 2, 3, 4$, the moments $\mu_k = \mathbb{E}(Z^k)$ are given by:

- $\mu_1 = \delta \, \mathbb{E}(T_{(-\bar{\tau}, \infty)}), \quad \nu > 1$
- $\mu_2 = \dfrac{\nu}{\nu - 1}(1 - \delta^2)\left(1 + \dfrac{1}{\nu} \mathbb{E}(T^2_{(-\bar{\tau}, \infty)}) \right) + \delta^2 \, \mathbb{E}(T^2_{(-\bar{\tau}, \infty)}), \quad \nu > 2$
- $\mu_3 = \dfrac{3\nu}{\nu - 1} \delta(1 - \delta^2) \left( \mathbb{E}(T_{(-\bar{\tau}, \infty)}) + \dfrac{1}{\nu} \mathbb{E}(T^3_{(-\bar{\tau}, \infty)}) \right) + \delta^3 \, \mathbb{E}(T^3_{(-\bar{\tau}, \infty)}), \quad \nu > 3$
- $\mu_4 = \dfrac{3\nu^2}{(\nu - 1)(\nu - 3)}(1 - \delta^2)^2 \left(1 + \dfrac{2}{\nu} \mathbb{E}(T^2_{(-\bar{\tau}, \infty)}) + \dfrac{1}{\nu^2} \mathbb{E}(T^4_{(-\bar{\tau}, \infty)}) \right)$  
  $\quad \quad \quad + \dfrac{6\nu}{\nu - 1} \delta^2(1 - \delta^2) \left( \mathbb{E}(T^2_{(-\bar{\tau}, \infty)}) + \dfrac{1}{\nu} \mathbb{E}(T^4_{(-\bar{\tau}, \infty)}) \right) + \delta^4 \, \mathbb{E}(T^4_{(-\bar{\tau}, \infty)}), \quad \nu > 4$

여기에서 

- $\mathbb{E}(T_{(-\bar{\tau}, \infty)}) = \dfrac{\nu}{\nu - 1} \left(1 + \dfrac{\bar{\tau}^2}{\nu} \right) t(\bar{\tau}; \nu) / T(\bar{\tau}; \nu), \quad \nu > 1$
- $\mathbb{E}(T^2_{(-\bar{\tau}, \infty)}) = \dfrac{\nu}{\nu - 2} \dfrac{T(\bar{\tau}_2; \nu - 2)}{T(\bar{\tau}; \nu)} - \bar{\tau} \, \mathbb{E}(T_{(-\bar{\tau}, \infty)}), \quad \nu > 2$
- $\mathbb{E}(T^3_{(-\bar{\tau}, \infty)}) = \dfrac{2\nu^2}{(\nu - 1)(\nu - 3)} \left(1 + \dfrac{\bar{\tau}^2}{\nu} \right)^2 t(\bar{\tau}; \nu) / T(\bar{\tau}; \nu) + \bar{\tau}^2 \, \mathbb{E}(T_{(-\bar{\tau}, \infty)}), \quad \nu > 3$
- $\mathbb{E}(T^4_{(-\bar{\tau}, \infty)}) = \dfrac{3\nu^2}{(\nu - 2)(\nu - 4)} \dfrac{T(\bar{\tau}_4; \nu - 4)}{T(\bar{\tau}; \nu)} - \dfrac{3}{2} \bar{\tau} \, \mathbb{E}(T^3_{(-\bar{\tau}, \infty)}) + \dfrac{1}{2} \bar{\tau}^3 \, \mathbb{E}(T_{(-\bar{\tau}, \infty)}), \quad \nu > 4$

- $\delta = \dfrac{\lambda}{\sqrt{1 + \lambda^2}}$
- $\bar{\tau} = \dfrac{\tau}{\sqrt{1 + \lambda^2}}$
- $\bar{\tau}_r = \sqrt{\dfrac{\nu + r}{\nu}} \bar{\tau}, \quad \nu + r > 0$
- $t(\cdot; \nu)$: univariate $t$ density with $\nu$ degrees of freedom  
- $T(\cdot; \nu)$: univariate $t$ CDF with $\nu$ degrees of freedom

```python
import numpy as np
from scipy.stats import t
import warnings

def delta_lambda(lambda_):
    return lambda_ / np.sqrt(1 + lambda_**2)

def tau_bar(tau, lambda_):
    return tau / np.sqrt(1 + lambda_**2)

def E_T1(tau_bar, nu):
    return (nu / (nu - 1)) * (1 + tau_bar**2 / nu) * t.pdf(tau_bar, nu) / t.cdf(tau_bar, nu)

def E_T2(tau_bar, nu):
    tau_bar_2 = np.sqrt((nu + 2) / nu) * tau_bar
    return (nu / (nu - 2)) * t.cdf(tau_bar_2, nu - 2) / t.cdf(tau_bar, nu) - tau_bar * E_T1(tau_bar, nu)

def E_T3(tau_bar, nu):
    return (2 * nu**2 / ((nu - 1) * (nu - 3))) * (1 + tau_bar**2 / nu)**2 * t.pdf(tau_bar, nu) / t.cdf(tau_bar, nu) + tau_bar**2 * E_T1(tau_bar, nu)

def E_T4(tau_bar, nu):
    tau_bar_4 = np.sqrt((nu + 4) / nu) * tau_bar
    term1 = (3 * nu**2 / ((nu - 2) * (nu - 4))) * t.cdf(tau_bar_4, nu - 4) / t.cdf(tau_bar, nu)
    term2 = (3 / 2) * tau_bar * E_T3(tau_bar, nu)
    term3 = (1 / 2) * tau_bar**3 * E_T1(tau_bar, nu)
    return term1 - term2 + term3

def EST_mean(xi, omega, lambda_, nu, tau):
    if nu <= 1:
        warnings.warn("Mean is undefined for nu <= 1.")
        return None
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)
    mu1 = delta * E_T1(tau_b, nu)
    return omega * mu1 + xi

def EST_var(xi, omega, lambda_, nu, tau):
    if nu <= 2:
        warnings.warn("Variance is undefined for nu <= 2.")
        return None
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)
    mu1 = delta * E_T1(tau_b, nu)
    mu2 = (nu / (nu - 1)) * (1 - delta**2) * (1 + E_T2(tau_b, nu) / nu) + delta**2 * E_T2(tau_b, nu)
    VarZ = mu2 - mu1**2
    return omega**2 * VarZ

def EST_skew(xi, omega, lambda_, nu, tau):
    if nu <= 3:
        warnings.warn("Skewness is undefined for nu <= 3.")
        return None
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)
    mu1 = delta * E_T1(tau_b, nu)
    mu2 = (nu / (nu - 1)) * (1 - delta**2) * (1 + E_T2(tau_b, nu) / nu) + delta**2 * E_T2(tau_b, nu)
    mu3 = (3 * nu / (nu - 1)) * delta * (1 - delta**2) * (E_T1(tau_b, nu) + E_T3(tau_b, nu) / nu) + delta**3 * E_T3(tau_b, nu)
    VarZ = mu2 - mu1**2
    return (mu3 - 3 * mu1 * VarZ - mu1**3) / VarZ**1.5

def EST_kurt(xi, omega, lambda_, nu, tau):
    if nu <= 4:
        warnings.warn("Kurtosis is undefined for nu <= 4.")
        return None
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)
    mu1 = delta * E_T1(tau_b, nu)
    mu2 = (nu / (nu - 1)) * (1 - delta**2) * (1 + E_T2(tau_b, nu) / nu) + delta**2 * E_T2(tau_b, nu)
    mu3 = (3 * nu / (nu - 1)) * delta * (1 - delta**2) * (E_T1(tau_b, nu) + E_T3(tau_b, nu) / nu) + delta**3 * E_T3(tau_b, nu)
    mu4 = (3 * nu**2 / ((nu - 1) * (nu - 3))) * (1 - delta**2)**2 * \
          (1 + 2 * E_T2(tau_b, nu) / nu + E_T4(tau_b, nu) / nu**2) + \
          (6 * nu / (nu - 1)) * delta**2 * (1 - delta**2) * \
          (E_T2(tau_b, nu) + E_T4(tau_b, nu) / nu) + \
          delta**4 * E_T4(tau_b, nu)
    VarZ = mu2 - mu1**2
    return (mu4 - 4 * mu3 * mu1 + 6 * mu2 * mu1**2 - 3 * mu1**4) / VarZ**2
```

### CDF of the EST

$\it EST$ 분포의 누적분포함수(CDF)는 식 (7) 또는 (9)를 이용하여 쉽게 계산할 수 있다. 만약 $Y \sim \mathrm{EST}_p(\xi, \Omega, \lambda, \nu, \tau)$라면, 다음이 성립한다:

$$
P(Y \leq y) = \frac{1}{T_1(\bar{\tau}; \nu)} \, T_{p+1}
\left(
\begin{bmatrix}
z \\
\bar{\tau}
\end{bmatrix}
;
\begin{bmatrix}
\bar{\Omega} & -\delta \\
-\delta^\top & 1
\end{bmatrix},
\nu + 1
\right),
$$

여기서:

- $z = \Omega^{-1}(y - \xi)$  
- $\bar{\tau} = \dfrac{\tau}{\sqrt{1 + \lambda^\top \lambda}}$  
- $\delta = \dfrac{\lambda}{\sqrt{1 + \lambda^\top \lambda}}$  
- $\bar{\Omega} = \Omega$ (혹은 관련된 변환에 따라 다를 수 있음)  
- $T_k(\cdot; \Sigma, \nu)$는 자유도 $\nu$와 공분산 $\Sigma$를 갖는 $k$-변량 $t$ 분포의 누적분포함수

### 단변량 EST 분포의 누적분포함수 (CDF)

만약 $ Y \sim \mathrm{EST}_1(\xi, \omega^2, \lambda, \nu, \tau) $라면, 그 누적분포함수는 다음과 같다:

$$
P(Y \leq y) = \frac{1}{T_1(\bar{\tau}; \nu)} \, T_2
\left(
\begin{bmatrix}
z \\
\bar{\tau}
\end{bmatrix}
;
\begin{bmatrix}
1 & -\delta \\
-\delta & 1
\end{bmatrix},
\nu + 1
\right),
$$

여기서:

- $ z = \dfrac{y - \xi}{\omega} $
- $ \bar{\tau} = \dfrac{\tau}{\sqrt{1 + \lambda^2}} $
- $ \delta = \dfrac{\lambda}{\sqrt{1 + \lambda^2}} $
- $ T_2(\cdot; \Sigma, \nu) $는 자유도 $ \nu $와 공분산 행렬 $ \Sigma $를 갖는 2변량 Student's $ t $ 누적분포함수

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import t, multivariate_t

# EST CDF function
def est_cdf(x, xi, omega, lam, nu, tau):
    delta = lam / np.sqrt(1 + lam**2)
    tau_bar = tau / np.sqrt(1 + lam**2)
    z = (x - xi) / omega

    # Bivariate t distribution setup
    mean = np.zeros(2)
    cov = np.array([[1, -delta], [-delta, 1]])

    # Denominator: marginal t CDF
    T_tau_bar = t.cdf(tau_bar, df=nu)

    # Numerator: bivariate t CDF
    x_vec = np.array([z, tau_bar])
    numerator = multivariate_t.cdf(x_vec, df=nu+1, loc=mean, shape=cov)

    return numerator / T_tau_bar

# Parameters
xi, omega, lam, nu, tau = 0, 1, 2, 3, 0

# y range
y_vals = np.linspace(-4, 4, 500)
cdf_vals = [est_cdf(y, xi, omega, lam, nu, tau) for y in y_vals]

# Plot
plt.figure(figsize=(8, 5))
plt.plot(y_vals, cdf_vals, label=f'$EST$({xi}, {omega**2}, {lam}, {nu}, {tau})')
plt.title('CDF of $EST$ Distribution')
plt.xlabel('$y$')
plt.ylabel('$P(Y ≤ y)$')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

### 로그가능도 함수(Log-likelihood function)

$\it EST$ 분포의 로그가능도 함수는 다음과 같이 주어진다:

$$
\ln L = \sum_{i=1}^n \ln f(x_i) = \sum_{i=1}^n \Bigg[ - \ln T\left(\frac{\tau}{\sqrt{1 + \lambda^2}}; \nu \right) 
+ \ln t(x_i; \xi, \omega, \nu)
+ \ln T \left\{ \left(\lambda z_i + \tau \right) \left(\frac{\nu + 1}{\nu + z_i^2}\right)^{1/2} ; \nu + 1 \right\}
\Bigg]
$$
여기서
$$
\ln t(x_i; \xi, \omega, \nu) = \ln \Gamma \left(\frac{\nu + 1}{2}\right) - \frac{1}{2}\ln(\pi \nu) - \ln\omega - \ln\Gamma\left(\frac{\nu}{2}\right) - \frac{\nu + 1}{2}\ln\left( 1 + \frac{z_i^2}{\nu} \right)
$$
여기서 $z_i = (x_i - \xi)/\omega$이고

```python
from scipy.special import gamma, gammaln

def ll_EST_e(x, xi, omega, lam, nu, tau, weight=None):

    omega = np.exp(omega)
    nu = np.exp(nu)

    z = (x - xi) / omega
    bar_tau = tau / np.sqrt(1 + lam**2)

    # log t(x; xi, omega, nu)
    log_t_density = (
        gammaln((nu + 1) / 2)
        - 0.5 * np.log(np.pi * nu)
        - np.log(omega)
        - gammaln(nu / 2)
        - (nu + 1) / 2 * np.log(1 + (z ** 2) / nu)
    )

    # CDF terms
    T_bar_tau = t.cdf(bar_tau, df=nu)
    T_skew_part = t.cdf((lam * z + tau) * np.sqrt((nu + 1) / (nu + z ** 2)), df=nu + 1)

    ln_pdf_vals = -np.log(T_bar_tau) + log_t_density + np.log(T_skew_part)

    # 가중치 적용 (있다면)
    if weight is not None:
        ln_pdf_vals *= weight
    
    log_lik_val = ln_pdf_vals.sum()
    
    return log_lik_val
```

```python
def obj_EST_e(params, *args):
    xi, omega, lam, nu, tau = params

    if len(args) == 2:
        x, weight = args
        log_lik_val = ll_EST_e(x, xi, omega, lam, nu, tau, weight=weight)
    else:
        x = args[0]
        log_lik_val = ll_EST_e(x, xi, omega, lam, nu, tau)

    return -log_lik_val
```

```python
import numpy as np
from scipy.stats import t
from scipy.special import gammaln

def ll_bi_EST_e(x1, x2, xi1, xi2, omega1, omega2, lam1, lam2, nu, tau, rho, weight=None):
    # 변환
    omega1 = np.exp(omega1)
    omega2 = np.exp(omega2)
    nu = np.exp(nu)

    # 정규화된 변수 z1, z2
    z1 = (x1 - xi1) / omega1
    z2 = (x2 - xi2) / omega2

    Q = (z1**2 - 2 * rho * z1 * z2 + z2**2) / (1 - rho**2)
    log_t_density = -(np.log(2 * np.pi) + np.log(omega1) + np.log(omega2) + 0.5 * np.log(1 - rho**2))\
                    - ((nu + 2) / 2) * np.log(1 + Q / nu)
    bar_tau = tau / np.sqrt(1 + lam1**2 + 2 * rho * lam1 * lam2 + lam2**2)
    skew_arg = (lam1 * z1 + lam2 * z2 + tau) * np.sqrt((nu + 2) / (nu + Q))

    # 로그우도 계산
    ln_pdf_vals = -np.log(t.cdf(bar_tau, df=nu)) + log_t_density + np.log(t.cdf(skew_arg, df=nu + 2))
    
    # 가중치 적용(있다면)
    if weight is not None:
        ln_pdf_vals *= weight

    log_lik_val = np.sum(ln_pdf_vals)

    return log_lik_val
```

```python
def obj_bi_EST_e(params, *args):
    xi1, xi2, omega1, omega2, lam1, lam2, nu, tau, rho = params

    if len(args) == 2:
        x, weight = args
        log_lik_val = ll_bi_EST_e(x1, x2, xi1, xi2, omega1, omega2, lam1, lam2, nu, tau, rho, weight=weight)
    else:
        x = args[0]
        log_lik_val = ll_bi_EST_e(x1, x2, xi1, xi2, omega1, omega2, lam1, lam2, nu, tau, rho)

    return -log_lik_val
```

### 주요 통계량

```python
import numpy as np
from scipy.stats import moment

def weighted_mean(x, weights):
    return np.sum(x * weights) / np.sum(weights)

def weighted_moment(data, weights, order):
    mean = weighted_mean(data, weights)
    return np.sum(weights * (data - mean) ** order) / np.sum(weights)

def weighted_std(data, weights):
    mean = weighted_mean(data, weights)
    variance = np.sum(weights * ((data - mean) ** 2)) / np.sum(weights)
    return np.sqrt(variance)

def weighted_skewness(data, weights):
    m2 = weighted_moment(data, weights, 2)
    m3 = weighted_moment(data, weights, 3)
    skewness = (np.sqrt(np.sum(weights)) * m3) / (m2 ** 1.5)
    return skewness

def weighted_kurtosis(data, weights):
    m2 = weighted_moment(data, weights, 2)
    m4 = weighted_moment(data, weights, 4)
    kurtosis = (np.sum(weights) * m4) / (m2 ** 2)
    return kurtosis
```

### SU

**PDF**

```python
def SU_pdf(x, m, s, lambda_, theta):

    J = 1/(theta*np.sqrt((x-m)**2+s**2))
    z = (np.arcsinh((x - m)/s) - lambda_)/theta
    pdf_vals = J*st.norm.pdf(z)
    
    return pdf_vals
```

```python
def ll_SU_e(x, m, s, lambda_, theta, weight=None):
    s = np.exp(s)
    theta = np.exp(theta)
    
    J = 1 / (theta * np.sqrt((x - m) ** 2 + s ** 2))
    z = (np.arcsinh((x - m) / s) - lambda_) / theta

    ln_pdf_vals = np.log(J) - 0.5 * np.log(2 * np.pi) - 0.5 * z ** 2

    if weight is not None:
        ln_pdf_vals *= weight

    log_lik_val = ln_pdf_vals.sum()
    
    return log_lik_val
```

**MLE Objective Function**

```python
def obj_SU_e(params, *args):
    m, s, lambda_, theta = params

    if len(args) == 2:
        x, weight = args
    else:
        (x,) = args
        weight = None

    log_lik_val = ll_SU_e(x, m, s, lambda_, theta, weight=weight)
    neg_log_lik_val = -log_lik_val

    return neg_log_lik_val
```

**Mean and Variance**

```python
def SU_mean(m, s, lambda_, theta):
    
    w = np.exp(theta**2)
    mean = m + s*w**(0.5)*np.sinh(lambda_)

    return mean

def SU_var(m, s, lambda_, theta):
    
    w = np.exp(theta**2)
    variance = 0.5*(s**2)*(w-1)*(w*np.cosh(2*lambda_)+1) 

    return variance

def SU_skew(m, s, lambda_, theta):
    
    w = np.exp(theta**2)
    sigma = (0.5*(s**2)*(w-1)*(w*np.cosh(2*lambda_)+1))**(0.5)
    mu_3 = 0.25*(s**3)*w**(0.5)*(w-1)**2*(w*(w+2)*np.sinh(3*lambda_)+3*np.sinh(lambda_))
    skewness = mu_3/(sigma**3)
        
    return skewness

def SU_kurt(m, s, lambda_, theta):
    
    w = np.exp(theta**2)
    sigma = (0.5*(s**2)*(w-1)*(w*np.cosh(2*lambda_)+1))**(0.5)
    mu_4 = 0.125*(s**4)*(w-1)**2*((w**2)*(w**4+2*w**3+3*w**2-3)*np.cosh(4*lambda_)
                                  + 4*(w**2)*(w+2)*np.cosh(2*lambda_) 
                                  + 3*(2*w+1))
    kurtosis = mu_4/(sigma**4)
        
    return kurtosis
```

**CDF and PPF**

- PPF stands for Probability Point Function. The PPF is the inverse of the CDF (cumulative distribution function). It returns the exact point where the probability of everything to the left is equal to y.

```python
def SU_cdf(x, m, s, lambda_, theta):
    
    z = (np.arcsinh((x - m)/s) - lambda_)/theta
    prob  = st.norm.cdf(z) 

    return prob

def SU_ppf(prob, m, s, lambda_, theta):
    
    x  = np.sinh(st.norm.ppf(prob)*theta + lambda_)*s + m

    return x
```

### Log Likelihood Functions

**일변량**

```python
def ll_SU_e(x, m, s, lambda_, theta, weight=None):
    s = np.exp(s)
    theta = np.exp(theta)
    
    J = 1 / (theta * np.sqrt((x - m) ** 2 + s ** 2))
    z = (np.arcsinh((x - m) / s) - lambda_) / theta

    ln_pdf_vals = np.log(J) - 0.5 * np.log(2 * np.pi) - 0.5 * z ** 2

    if weight is not None:
        ln_pdf_vals *= weight

    log_lik_val = ln_pdf_vals.sum()
    
    return log_lik_val
```

**이변량**

```python
def ll_bi_SU_e(x1, x2, m1, m2, s1, s2, lambda1, lambda2, theta1, theta2, r, weight=None):
    s1 = np.exp(s1)
    theta1 = np.exp(theta1)
    s2 = np.exp(s2)
    theta2 = np.exp(theta2)

    # Jacobian
    J = 1 / ((theta1 * np.sqrt((x1 - m1) ** 2 + s1 ** 2)) * (theta2 * np.sqrt((x2 - m2) ** 2 + s2 ** 2)))
    
    # z-transformed variables
    z1 = (np.arcsinh((x1 - m1) / s1) - lambda1) / theta1
    z2 = (np.arcsinh((x2 - m2) / s2) - lambda2) / theta2

    # Log density
    ln_pdf_vals = (
        np.log(J)
        - np.log(2 * np.pi)
        - 0.5 * np.log(1 - r ** 2)
        - 0.5 / (1 - r ** 2) * (z1 ** 2 + z2 ** 2 - 2 * r * z1 * z2)
    )

    if weight is not None:
        ln_pdf_vals *= weight

    log_lik_val = ln_pdf_vals.sum()

    return log_lik_val
```

### MLE Objective Functions

**일변량**

```python
def obj_SU_e(params, *args):
    m, s, lambda_, theta = params

    if len(args) == 2:
        x, weight = args
    else:
        (x,) = args
        weight = None

    log_lik_val = ll_SU_e(x, m, s, lambda_, theta, weight=weight)
    neg_log_lik_val = -log_lik_val

    return neg_log_lik_val
```

**이변량**

```python
def obj_bi_SU_e(params, *args):
    m1, s1, lambda1, theta1, m2, s2, lambda2, theta2, r = params

    if len(args) == 3:
        x1, x2, weight = args
    else:
        x1, x2 = args
        weight = None

    log_lik_val = ll_bi_SU_e(x1, x2, m1, m2, s1, s2, lambda1, lambda2, theta1, theta2, r, weight=weight)
    neg_log_lik_val = -log_lik_val
    
    return neg_log_lik_val
```

```python
m1_init = 0
s1_init = 0
lambda1_init = 0
theta1_init = 0
m2_init = 0
s2_init = 0
lambda2_init = 0
theta2_init = 0
r = 0
params_init = np.array([m1_init, s1_init, lambda1_init, theta1_init,
                       m2_init, s2_init, lambda2_init, theta2_init, r])
mle_args = (x1,x2,w)
res_bi_SU = opt.minimize(obj_bi_SU_e, params_init, args=mle_args,
                         method='L-BFGS-B',
                         bounds=[(None, None)] * 8 + [(-.99, .99)])

(m1_MLE, s1_MLE, lambda1_MLE, theta1_MLE,
 m2_MLE, s2_MLE, lambda2_MLE, theta2_MLE, r_MLE)=res_bi_SU.x

s1_MLE = np.exp(s1_MLE)
theta1_MLE = np.exp(theta1_MLE)
s2_MLE = np.exp(s2_MLE)
theta2_MLE = np.exp(theta2_MLE)

result = {'m1': [m1_MLE], 's1': [s1_MLE], 
          'lambda1': [lambda1_MLE], 'theta1': [theta1_MLE],
          'm2': [m2_MLE], 's2': [s2_MLE],
          'lambda2': [lambda2_MLE], 'theta2': [theta2_MLE], 'r': [r_MLE]}
df = pd.DataFrame(result)
df.T
```
