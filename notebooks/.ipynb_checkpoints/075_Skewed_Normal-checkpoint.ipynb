## 1. Functions

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy.optimize as opt
import scipy.stats as st
from scipy.special import gamma, gammaln
import warnings
```

### EST

#### PDF CDF

```python
def ESN_pdf(x, xi, omega, lam, tau):
    # 표준화된 값
    z = (x - xi) / omega

    # 정규화 상수 (denominator)
    denom_cdf = st.norm.cdf(tau / np.sqrt(1 + lam**2))

    # 정규분포 밀도
    normal_density = st.norm.pdf(x, loc=xi, scale=omega)

    # 비대칭성을 주는 누적분포 함수 인자
    arg = lam * z + tau
    cdf_term = st.norm.cdf(arg)

    # ESN 밀도 함수
    return (1 / denom_cdf) * normal_density * cdf_term
```

```python
# EST CDF function
def est_cdf(x, xi, omega, lam, nu, tau):
    delta = lam / np.sqrt(1 + lam**2)
    tau_bar = tau / np.sqrt(1 + lam**2)
    z = (x - xi) / omega

    # Bivariate t distribution setup
    mean = np.zeros(2)
    cov = np.array([[1, -delta], [-delta, 1]])

    # Denominator: marginal t CDF
    T_tau_bar = st.t.cdf(tau_bar, df=nu)

    # Numerator: bivariate t CDF
    x_vec = np.array([z, tau_bar])
    numerator = st.multivariate_t.cdf(x_vec, df=nu+1, loc=mean, shape=cov)

    return numerator / T_tau_bar
```

#### Moments

```python
def delta_lambda(lambda_):
    return lambda_ / np.sqrt(1 + lambda_**2)

def tau_bar(tau, lambda_):
    return tau / np.sqrt(1 + lambda_**2)

def E_T1(tau_bar, nu):
    return (nu / (nu - 1)) * (1 + tau_bar**2 / nu) * st.t.pdf(tau_bar, nu) / st.t.cdf(tau_bar, nu)

def E_T2(tau_bar, nu):
    tau_bar_2 = np.sqrt((nu + 2) / nu) * tau_bar
    return ((nu / (nu - 2)) * st.t.cdf(tau_bar_2, nu - 2) / st.t.cdf(tau_bar, nu) 
             - tau_bar * E_T1(tau_bar, nu))

def E_T3(tau_bar, nu):
    return ((2 * nu**2 / ((nu - 1) * (nu - 3))) * (1 + tau_bar**2 / nu)**2 * st.t.pdf(tau_bar, nu)/
            st.t.cdf(tau_bar, nu) + tau_bar**2 * E_T1(tau_bar, nu))

def E_T4(tau_bar, nu):
    tau_bar_4 = np.sqrt((nu + 4) / nu) * tau_bar
    term1 = (3 * nu**2 / ((nu - 2)*(nu - 4))) * st.t.cdf(tau_bar_4, nu - 4) / st.t.cdf(tau_bar, nu)
    term2 = (3 / 2) * tau_bar * E_T3(tau_bar, nu)
    term3 = (1 / 2) * tau_bar**3 * E_T1(tau_bar, nu)
    return term1 - term2 + term3

def EST_mean(xi, omega, lambda_, nu, tau):
    if nu <= 1:
        warnings.warn("Mean is undefined for nu <= 1.")
        return None
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)
    mu1 = delta * E_T1(tau_b, nu)
    return omega * mu1 + xi

def EST_var(xi, omega, lambda_, nu, tau):
    if nu <= 2:
        warnings.warn("Variance is undefined for nu <= 2.")
        return None
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)
    mu1 = delta * E_T1(tau_b, nu)
    mu2 = (nu / (nu - 1)) * (1 - delta**2) * (1 + E_T2(tau_b, nu) / nu) \
          + delta**2 * E_T2(tau_b, nu)
    VarZ = mu2 - mu1**2
    return omega**2 * VarZ

def EST_skew(xi, omega, lambda_, nu, tau):
    if nu <= 3:
        warnings.warn("Skewness is undefined for nu <= 3.")
        return None
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)
    mu1 = delta * E_T1(tau_b, nu)
    mu2 = (nu / (nu - 1)) * (1 - delta**2) * (1 + E_T2(tau_b, nu) / nu) \
          + delta**2 * E_T2(tau_b, nu)
    mu3 = (3 * nu / (nu - 1)) * delta * (1 - delta**2) * \
          (E_T1(tau_b, nu) + E_T3(tau_b, nu) / nu) + delta**3 * E_T3(tau_b, nu)
    VarZ = mu2 - mu1**2
    return (mu3 - 3 * mu1 * VarZ - mu1**3) / VarZ**1.5

def EST_kurt(xi, omega, lambda_, nu, tau):
    if nu <= 4:
        warnings.warn("Kurtosis is undefined for nu <= 4.")
        return None
    delta = delta_lambda(lambda_)
    tau_b = tau_bar(tau, lambda_)
    mu1 = delta * E_T1(tau_b, nu)
    mu2 = (nu / (nu - 1)) * (1 - delta**2) * (1 + E_T2(tau_b, nu) / nu) \
          + delta**2 * E_T2(tau_b, nu)
    mu3 = (3 * nu / (nu - 1)) * delta * (1 - delta**2) * \
          (E_T1(tau_b, nu) + E_T3(tau_b, nu) / nu) + delta**3 * E_T3(tau_b, nu)
    mu4 = (3 * nu**2 / ((nu - 1) * (nu - 3))) * (1 - delta**2)**2 * \
          (1 + 2 * E_T2(tau_b, nu) / nu + E_T4(tau_b, nu) / nu**2) + \
          (6 * nu / (nu - 1)) * delta**2 * (1 - delta**2) * \
          (E_T2(tau_b, nu) + E_T4(tau_b, nu) / nu) + \
          delta**4 * E_T4(tau_b, nu)
    VarZ = mu2 - mu1**2
    return (mu4 - 4 * mu3 * mu1 + 6 * mu2 * mu1**2 - 3 * mu1**4) / VarZ**2
```

```python
def zeta1(a):
    """E[N_{(-a, ∞)}] = φ(a) / Φ(a)"""
    return st.norm.pdf(a) / st.norm.cdf(a)

def ESN_moments(lambda_, tau):
    delta = lambda_ / np.sqrt(1 + lambda_**2)
    bar_tau = tau / np.sqrt(1 + lambda_**2)
    ζ = zeta1(bar_tau)

    E1 = ζ
    E2 = 1 - bar_tau * ζ
    E3 = (2 + bar_tau**2) * ζ
    E4 = 3 - (3 * bar_tau + bar_tau**3) * ζ

    μ1 = delta * E1
    μ2 = (1 - delta**2) + delta**2 * E2
    μ3 = 3 * delta * (1 - delta**2) * E1 + delta**3 * E3
    μ4 = (
        3 * (1 - delta**2)**2
        + 6 * delta**2 * (1 - delta**2) * E2
        + delta**4 * E4
    )

    return μ1, μ2, μ3, μ4

def ESN_mean(xi, omega, lambda_, tau):
    μ1, _, _, _ = ESN_moments(lambda_, tau)
    return xi + omega * μ1

def ESN_std(xi, omega, lambda_, tau):
    μ1, μ2, _, _ = ESN_moments(lambda_, tau)
    return omega * np.sqrt(μ2 - μ1**2)

def ESN_skew(xi, omega, lambda_, tau):
    μ1, μ2, μ3, _ = ESN_moments(lambda_, tau)
    σ = np.sqrt(μ2 - μ1**2)
    return (μ3 - 3 * μ1 * σ**2 - μ1**3) / σ**3

def ESN_kurt(xi, omega, lambda_, tau):
    μ1, μ2, μ3, μ4 = ESN_moments(lambda_, tau)
    σ2 = μ2 - μ1**2
    return (μ4 - 4 * μ1 * μ3 + 6 * μ1**2 * μ2 - 3 * μ1**4) / σ2**2
```

#### Log Likelihood Functions

```python
def ll_ESN_e(x, xi, omega, lam, tau, weight=None):
    # omega는 로그 스케일로 주어짐
    omega = np.exp(omega)

    z = (x - xi) / omega
    bar_tau = tau / np.sqrt(1 + lam**2)

    # log phi(x; xi, omega)
    log_norm_density = st.norm.logpdf(x, loc=xi, scale=omega)

    # log Phi(lambda * z + tau)
    skew_arg = lam * z + tau
    log_skew_cdf = st.norm.logcdf(skew_arg)

    # log 정규화 상수
    log_denom_cdf = np.log(st.norm.cdf(bar_tau))

    # 전체 로그 우도
    ln_pdf_vals = log_norm_density + log_skew_cdf - log_denom_cdf

    if weight is not None:
        ln_pdf_vals *= weight

    return ln_pdf_vals.sum()
```

```python
def ll_bi_ESN_e(x1, x2, xi1, xi2, omega1, omega2, lam1, lam2, tau, rho, weight=None):
    # 로그 스케일에서 실제 scale로 변환
    omega1 = np.exp(omega1)
    omega2 = np.exp(omega2)

    # 정규화된 변수
    z1 = (x1 - xi1) / omega1
    z2 = (x2 - xi2) / omega2

    # 공분산 행렬의 정규화 상수
    Q = (z1**2 - 2 * rho * z1 * z2 + z2**2) / (1 - rho**2)

    # log phi_2(x; xi, Omega)
    log_norm_density = -np.log(2 * np.pi) - np.log(omega1) - np.log(omega2) \
                       - 0.5 * np.log(1 - rho**2) - 0.5 * Q

    # 정규화 상수
    bar_tau = tau / np.sqrt(1 + lam1**2 + 2 * rho * lam1 * lam2 + lam2**2)

    # 비대칭성 반영 cdf
    skew_arg = lam1 * z1 + lam2 * z2 + tau
    log_skew_cdf = np.log(st.norm.cdf(skew_arg))

    # 로그 정규화 상수
    log_denom_cdf = np.log(st.norm.cdf(bar_tau))

    # 로그-밀도
    ln_pdf_vals = log_norm_density + log_skew_cdf - log_denom_cdf

    # 가중치 적용
    if weight is not None:
        ln_pdf_vals *= weight

    return np.sum(ln_pdf_vals)
```

#### MLE Objective Functions

```python
def obj_ESN_e(params, *args):
    xi, omega, lam, tau = params

    if len(args) == 2:
        x, weight = args
    else:
        x = args
        weight = None

    log_lik_val = ll_ESN_e(x, xi, omega, lam, tau, weight=weight)
    neg_log_lik_val = -log_lik_val

    return neg_log_lik_val
```

```python
def obj_bi_ESN_e(params, *args):
    xi1, xi2, omega1, omega2, lam1, lam2, tau, rho = params

    if len(args) == 3:
        x1, x2, weight = args
    else:
        x1, x2 = args
        weight = None

    log_lik_val = ll_bi_ESN_e(x1, x2, xi1, xi2, omega1, omega2, lam1, lam2, tau, rho, weight=weight)
    neg_log_lik_val = -log_lik_val
    
    return neg_log_lik_val
```

### SU

#### PDFs

```python
def SU_pdf(x, m, s, lambda_, theta):

    J = 1/(theta*np.sqrt((x-m)**2+s**2))
    z = (np.arcsinh((x - m)/s) - lambda_)/theta
    pdf_vals = J*st.norm.pdf(z)
    
    return pdf_vals
```

```python
def SU_bi_pdf(x1, x2, m1, m2, s1, s2, lambda1, lambda2, theta1, theta2, r):

    J = 1/( (theta1*np.sqrt((x1-m1)**2+s1**2))*(theta2*np.sqrt((x2-m1)**2+s2**2)) )
    z1 = (np.arcsinh((x1 - m1)/s1) - lambda1)/theta1
    z2 = (np.arcsinh((x2 - m2)/s2) - lambda2)/theta2
    pdf_vals = 1/(2*np.pi*np.sqrt(1-r**2))*J*np.exp(
        -1/(2*(1-r**2))*(z1**2+z2**2-2*r*z1*z2))
    
    return pdf_vals
```

```python
def SU_cond_pdf(x1, x2, m1, m2, s1, s2, lambda1, lambda2, theta1, theta2, r):
    '''
    Conditional PDF of x1 given x2
    '''    
    
    lambda1_star = lambda1 + r * theta1/theta2 * (np.arcsinh((x2 - m2) / s2) - lambda2)
    theta1_star = theta1 * (1 - r**2)**(1/2)

    J = 1/(theta1_star*np.sqrt((x1-m1)**2+s1**2))
    z = (np.arcsinh((x1 - m1)/s1) - lambda1_star)/theta1_star
    pdf_vals = J*st.norm.pdf(z)
    
    return pdf_vals
```

#### Log Likelihood Functions

```python
def ll_SU_e(x, m, s, lambda_, theta, weight=None):
    s = np.exp(s)
    theta = np.exp(theta)
    
    J = 1 / (theta * np.sqrt((x - m) ** 2 + s ** 2))
    z = (np.arcsinh((x - m) / s) - lambda_) / theta

    ln_pdf_vals = np.log(J) - 0.5 * np.log(2 * np.pi) - 0.5 * z ** 2

    if weight is not None:
        ln_pdf_vals *= weight

    log_lik_val = ln_pdf_vals.sum()
    
    return log_lik_val
```

```python
def ll_bi_SU_e(x1, x2, m1, m2, s1, s2, lambda1, lambda2, theta1, theta2, r, weight=None):
    s1 = np.exp(s1)
    theta1 = np.exp(theta1)
    s2 = np.exp(s2)
    theta2 = np.exp(theta2)

    # Jacobian
    J = 1 / ((theta1 * np.sqrt((x1 - m1) ** 2 + s1 ** 2)) * (theta2 * np.sqrt((x2 - m2) ** 2 + s2 ** 2)))
    
    # z-transformed variables
    z1 = (np.arcsinh((x1 - m1) / s1) - lambda1) / theta1
    z2 = (np.arcsinh((x2 - m2) / s2) - lambda2) / theta2

    # Log density
    ln_pdf_vals = (
        np.log(J)
        - np.log(2 * np.pi)
        - 0.5 * np.log(1 - r ** 2)
        - 0.5 / (1 - r ** 2) * (z1 ** 2 + z2 ** 2 - 2 * r * z1 * z2)
    )

    if weight is not None:
        ln_pdf_vals *= weight

    log_lik_val = ln_pdf_vals.sum()

    return log_lik_val
```

#### MLE Objective Functions

```python
def obj_SU_e(params, *args):
    m, s, lambda_, theta = params

    if len(args) == 2:
        x, weight = args
    else:
        (x,) = args
        weight = None

    log_lik_val = ll_SU_e(x, m, s, lambda_, theta, weight=weight)
    neg_log_lik_val = -log_lik_val

    return neg_log_lik_val
```

```python
def obj_bi_SU_e(params, *args):
    m1, s1, lambda1, theta1, m2, s2, lambda2, theta2, r = params

    if len(args) == 3:
        x1, x2, weight = args
    else:
        x1, x2 = args
        weight = None

    log_lik_val = ll_bi_SU_e(x1, x2, m1, m2, s1, s2, lambda1, lambda2, theta1, theta2, r, weight=weight)
    neg_log_lik_val = -log_lik_val
    
    return neg_log_lik_val
```

#### Moments

```python
def SU_mean(m, s, lambda_, theta):
    
    w = np.exp(theta**2)
    mean = m + s*w**(0.5)*np.sinh(lambda_)

    return mean

def SU_var(m, s, lambda_, theta):
    
    w = np.exp(theta**2)
    variance = 0.5*(s**2)*(w-1)*(w*np.cosh(2*lambda_)+1) 

    return variance

def SU_skew(m, s, lambda_, theta):
    
    w = np.exp(theta**2)
    sigma = (0.5*(s**2)*(w-1)*(w*np.cosh(2*lambda_)+1))**(0.5)
    mu_3 = 0.25*(s**3)*w**(0.5)*(w-1)**2*(w*(w+2)*np.sinh(3*lambda_)+3*np.sinh(lambda_))
    skewness = mu_3/(sigma**3)
        
    return skewness

def SU_kurt(m, s, lambda_, theta):
    
    w = np.exp(theta**2)
    sigma = (0.5*(s**2)*(w-1)*(w*np.cosh(2*lambda_)+1))**(0.5)
    mu_4 = 0.125*(s**4)*(w-1)**2*((w**2)*(w**4+2*w**3+3*w**2-3)*np.cosh(4*lambda_)
                                  + 4*(w**2)*(w+2)*np.cosh(2*lambda_) 
                                  + 3*(2*w+1))
    kurtosis = mu_4/(sigma**4)
        
    return kurtosis
```

```python
# Pearson Correlation

def SU_P_corr(lambda1, lambda2, theta1, theta2, r):
    
    w1 = np.exp(theta1**2)
    w2 = np.exp(theta2**2)
    sigma1 = np.sqrt(0.5*(w1-1)*(w1*np.cosh(2*lambda1)+1)) 
    sigma2 = np.sqrt(0.5*(w2-1)*(w2*np.cosh(2*lambda2)+1)) 
    rho = (
        np.exp(0.5*(theta1**2+theta2**2))/(sigma1*sigma2)*
        (0.5*np.exp(r*theta1*theta2)*np.cosh(lambda1+lambda2)
         -0.5*np.exp(-r*theta1*theta2)*np.cosh(lambda1-lambda2)
         -np.sinh(lambda1)*np.sinh(lambda2))
    )
    
    return rho

# Normalized Correlation

def SU_N_corr(lambda1, lambda2, theta1, theta2, rho):
    
    w1 = np.exp(theta1**2)
    w2 = np.exp(theta2**2)
    sigma1 = np.sqrt(0.5*(w1-1)*(w1*np.cosh(2*lambda1)+1)) 
    sigma2 = np.sqrt(0.5*(w2-1)*(w2*np.cosh(2*lambda2)+1))
    B = (rho*sigma1*sigma2*np.exp(-0.5*(theta1**2+theta2**2)) 
         + np.sinh(lambda1)*np.sinh(lambda2))

    r = 1/(theta1*theta2)*np.log(
        (B+np.sqrt(B**2+np.cosh(lambda1+lambda2)*np.cosh(lambda1-lambda2)))/
        np.cosh(lambda1+lambda2))
    
    return r
```

### Weighted Statistics

```python
def weighted_mean(x, weights=None):
    if weights is None:
        return np.mean(x)
    return np.sum(weights * x) / np.sum(weights)

def weighted_std(x, weights=None):
    mean = weighted_mean(x, weights)
    if weights is None:
        return np.std(x, ddof=0)
    mean_sq = np.sum(weights * x**2) / np.sum(weights)
    return np.sqrt(mean_sq - mean**2)

def weighted_skewness(x, weights=None):
    mean = weighted_mean(x, weights)
    std = weighted_std(x, weights)
    if weights is None:
        m3 = np.mean((x - mean) ** 3)
        return m3 / std**3
    m3 = np.sum(weights * (x - mean) ** 3) / np.sum(weights)
    return m3 / std**3

def weighted_kurtosis(x, weights=None):
    mean = weighted_mean(x, weights)
    std = weighted_std(x, weights)
    if weights is None:
        m4 = np.mean((x - mean) ** 4)
        return m4 / std**4
    m4 = np.sum(weights * (x - mean) ** 4) / np.sum(weights)
    return m4 / std**4
```

```python
def weighted_pearson_corr(x, y, weights=None):
    if weights is None:
        # 일반 Pearson 상관계수
        return np.corrcoef(x, y)[0, 1]
    
    # 가중 평균
    mean_x = np.average(x, weights=weights)
    mean_y = np.average(y, weights=weights)
    
    # 가중 공분산 및 분산
    cov_xy = np.average((x - mean_x) * (y - mean_y), weights=weights)
    var_x = np.average((x - mean_x) ** 2, weights=weights)
    var_y = np.average((y - mean_y) ** 2, weights=weights)
    
    # 가중 Pearson 상관계수
    corr_xy = cov_xy / np.sqrt(var_x * var_y)
    return corr_xy
```
